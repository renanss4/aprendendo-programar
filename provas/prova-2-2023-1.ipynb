### Questão 1

Desenvolva uma **função** que receba uma matriz quadrada (caso não seja, retorne uma
lista vazia) e **retorne uma lista com o maior elemento de cada diagonal acima da
principal**. A diagonal mais próxima da principal tem índice 1, e os índices das demais
aumentam de 1 em 1.

```
Parâmetro        Retorno        Parâmetro         Retorno
1  1 -4          [1, -4]        1 2 3 4 6         [9, 8, 7, 6]
2 -1 -2                         1 2 4 5 7
3 -2  1                         1 2 5 7 8
                                0 5 7 1 9
1 2 4            [2, 4]         1 2 5 7 8
2 1 2
3 1 1                           8 -3 1            []
                                4 -4 2
```

````python
def eh_quadrada(matriz):
    for linha in matriz:
        if len(linha) != len(matriz):
            return False
    return True


def maiores_das_diagonais(matriz):
    if not eh_quadrada(matriz):
        return []

    n = len(matriz)
    resultado = []
    k = 1
    while k < n:
        maior = matriz[0][k]
        linha = 1
        while linha + k < n:
            if matriz[linha][linha + k] > maior:
                maior = matriz[linha][linha + k]
            linha += 1
        resultado.append(maior)
        k += 1
    return resultado


print(maiores_das_diagonais([[1, 1, -4], [2, -1, -2], [3, -2, 1]]))
print(maiores_das_diagonais([[1, 2, 4], [2, 1, 2], [3, 1, 1]]))
print(maiores_das_diagonais([
    [1, 2, 3, 4, 6], [1, 2, 4, 5, 7], [1, 2, 5, 7, 8], [0, 5, 7, 1, 9], [1, 2, 5, 7, 8]]))
print(maiores_das_diagonais([[8, -3, 1], [4, -4, 2]]))
````

**2.2** Sobrescreva o método `__str__` da classe `Data` para retornar a data no formato

```python
class Data:
    def __init__(self, dia=1, mes=1, ano=2023):
        meses = {1: 31, 2: 28, 3: 31, 4: 30, 5: 31, 6: 30, 7: 31, 8: 31, 9: 30,
                 10: 31, 11: 30, 12: 31}
```

````python
class Data:
    def __init__(self, dia=1, mes=1, ano=2023):
        meses = {1: 31, 2: 28, 3: 31, 4: 30, 5: 31, 6: 30, 7: 31, 8: 31, 9: 30,
                 10: 31, 11: 30, 12: 31}

        valida = 1 <= mes <= 12 and ano > 1900
        if valida:                  # só consulta o dicionário se o mês existir
            valida = 0 < dia <= meses[mes]

        if valida:
            self.__dia = dia
            self.__mes = mes
            self.__ano = ano
        else:
            self.__dia = 26
            self.__mes = 6
            self.__ano = 2023

    def __str__(self):
        return f'{self.__dia}/{self.__mes}/{self.__ano}'


print(Data(3, 4, 2023), Data(31, 4, 2023), Data(10, 13, 2023), Data(1, 1, 1800))
````

A ordem das verificações importa: se o mês for 13, `meses[13]` causaria `KeyError`.
Por isso testamos o mês antes de consultar o dicionário.

### Questão 3

Construa a classe `Bolsista`, subclasse da classe `Estudante` (código fornecido na
célula de preparação). Além dos atributos herdados, ela tem o atributo **protegido**
`dataInicio`, que representa a data em que a bolsa começou.

- a) **Construtor:** tem parâmetros para inicializar todos os atributos da classe. O
  parâmetro `dataInicio` é opcional e tem como valor padrão um objeto da classe `Data`
  representando 26/6/2026. É obrigatório executar o construtor da superclasse (com
  `super`) para inicializar os atributos herdados.

```python
class Estudante:
    def __init__(self, mat, iaa):
        self.__mat = mat
        self.__iaa = iaa

    def obtemMatricula(self):
        return self.__mat

    def obtemIAA(self):
        return self.__iaa
```

````python
class Bolsista(Estudante):
    def __init__(self, mat, iaa, dataInicio=None):
        super().__init__(mat, iaa)
        if dataInicio is None:
            dataInicio = Data(26, 6, 2026)
        self._dataInicio = dataInicio


bolsista = Bolsista('2023001', 8.5)
print(bolsista.obtemMatricula(), bolsista.obtemIAA(), bolsista._dataInicio)
````

Por que `dataInicio=None` em vez de `dataInicio=Data(26, 6, 2026)`? Valores padrão são
criados **uma única vez**, quando a função é definida. Com `Data(...)` direto no
parâmetro, todos os bolsistas criados sem data compartilhariam o **mesmo** objeto. Usar
`None` e criar o objeto dentro do construtor evita esse problema. (Na prova, as duas
formas seriam aceitas.)

### Questão 4

Copie a classe `Cadastro` abaixo para a sua célula e acrescente o método
`obtemEstudantesIAA`:

- o método retorna uma lista com os objetos `Estudante` ou `Bolsista` cujo IAA é
  **maior** que o parâmetro `iaa` do método;
- os estudantes ficam no atributo `cadastro`, que é um dicionário. Observe no método
  `adicionaEst` qual é a chave e qual é o valor.

```python
class Cadastro:
    def __init__(self):
        self.__cadastro = {}

    def adicionaEst(self, est):
        mat = est.obtemMatricula()
        self.__cadastro[mat] = est
```

````python
class Cadastro:
    def __init__(self):
        self.__cadastro = {}

    def adicionaEst(self, est):
        mat = est.obtemMatricula()
        self.__cadastro[mat] = est

    def obtemEstudantesIAA(self, iaa):
        resultado = []
        for mat in self.__cadastro.keys():
            estudante = self.__cadastro[mat]
            if estudante.obtemIAA() > iaa:
                resultado.append(estudante)
        return resultado


cadastro = Cadastro()
cadastro.adicionaEst(Estudante('1', 6.0))
cadastro.adicionaEst(Bolsista('2', 8.5))
cadastro.adicionaEst(Estudante('3', 9.1))
for est in cadastro.obtemEstudantesIAA(7.0):
    print(est.obtemMatricula(), est.obtemIAA())
````
