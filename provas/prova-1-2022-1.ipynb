### Questão 1

Desenvolva um programa que leia um número inteiro positivo (não é necessário
verificar se é positivo) e mostre esse número **invertido**. Não use listas.

| Entrada | Saída |
| ------- | ----- |
| 123 | 321 |
| 7 | 7 |
| 98702345 | 54320789 |

````python
numero = int(input('Número: '))

invertido = 0
while numero > 0:
    ultimo_digito = numero % 10
    invertido = invertido * 10 + ultimo_digito
    numero = numero // 10

print(invertido)
````

A cada volta, tiramos o último dígito de `numero` (`% 10` e `// 10`) e o colocamos no
final de `invertido` (multiplicando por 10 e somando). Teste de mesa com 123:
`invertido` passa por 3, 32 e 321.

### Questão 2

Desenvolva um programa que leia um número inteiro maior do que 1 (não é necessário
verificar) e mostre uma figura como nos exemplos abaixo. A entrada determina a altura e
a largura da figura.

```
Entrada 2     Entrada 3     Entrada 5
 *              *               *
**             **              **
              ***             ***
                             ****
                            *****
```

````python
n = int(input('Entrada: '))

linha = 1
while linha <= n:
    espacos = 1
    while espacos <= n - linha:
        print(' ', end='')
        espacos += 1
    asteriscos = 1
    while asteriscos <= linha:
        print('*', end='')
        asteriscos += 1
    print()
    linha += 1
````

### Questão 3

Desenvolva um programa com a função da letra a) e as funcionalidades da letra b).

**a) Função `semRepetidos`:** recebe uma lista e retorna uma nova lista com os
elementos da lista recebida **sem repetição**. A função também retorna o **maior**
elemento dentre os distintos, como **último elemento** da lista de retorno. As duas
análises devem ser feitas em uma única estrutura de repetição, percorrendo a lista
recebida apenas uma vez.

| Argumento | Retorno |
| --------- | ------- |
| `[2, 3, 4, 1, 3, 1, 0]` | `[2, 3, 4, 1, 0, 4]` |
| `[0, 9, 2, 1]` | `[0, 9, 2, 1, 9]` |
| `[]` | `[]` |

**b) Execução:** não é necessário obter entrada do usuário. O programa deve criar uma
lista (como as dos exemplos), executar a função da letra a), guardar o retorno e
mostrá-lo como nos exemplos abaixo:

```
[2, 3, 4, 1, 3, 1, 0]  →  elementos sem repetição: 2, 3, 4, 1, 0
                          maior dentre os elementos: 4
[]                     →  elementos sem repetição: lista vazia
                          maior dentre os elementos: nenhum
```

````python
def semRepetidos(lista):
    resultado = []
    maior = 0
    for elemento in lista:
        # verifica se o elemento já está no resultado (sem usar o operador in)
        repetido = False
        indice = 0
        while indice < len(resultado):
            if resultado[indice] == elemento:
                repetido = True
            indice += 1

        if not repetido:
            if len(resultado) == 0 or elemento > maior:
                maior = elemento
            resultado.append(elemento)

    if len(resultado) > 0:
        resultado.append(maior)
    return resultado


def mostrar(lista):
    retorno = semRepetidos(lista)
    if len(retorno) == 0:
        print('elementos sem repetição: lista vazia')
        print('maior dentre os elementos: nenhum')
    else:
        print('elementos sem repetição: ', end='')
        indice = 0
        while indice < len(retorno) - 1:
            if indice > 0:
                print(', ', end='')
            print(retorno[indice], end='')
            indice += 1
        print()
        print('maior dentre os elementos:', retorno[len(retorno) - 1])


mostrar([2, 3, 4, 1, 3, 1, 0])
mostrar([0, 9, 2, 1])
mostrar([])
````

O maior elemento é atualizado no mesmo laço que monta a lista sem repetições, como a
questão exige. O laço interno não percorre a lista recebida, e sim a lista de
resultado.
