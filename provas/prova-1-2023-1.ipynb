### Questão 1

Desenvolva um programa que leia um número inteiro maior do que 1 (não é necessário
verificar) e mostre uma figura como nos exemplos abaixo. A entrada determina a altura e
a largura da figura.

```
Entrada 2     Entrada 3     Entrada 4
1***1         1*****1       1*******1
12*12         12***12       12*****12
              123*123       123***123
                            1234*1234
```

````python
n = int(input('Entrada: '))

linha = 1
while linha <= n:
    numero = 1
    while numero <= linha:
        print(numero, end='')
        numero += 1

    asteriscos = 1
    while asteriscos <= 2 * (n - linha) + 1:
        print('*', end='')
        asteriscos += 1

    numero = 1
    while numero <= linha:
        print(numero, end='')
        numero += 1

    print()
    linha += 1
````

Cada linha `l` tem: os números de 1 a `l`, depois `2 * (n - l) + 1` asteriscos e, de
novo, os números de 1 a `l`.

### Questão 2

Desenvolva um programa com as funções das letras a) e b).

**a) Função `quantidades`:** recebe uma lista e um número e retorna quantas vezes o
número aparece na lista. A função deve percorrer a lista apenas uma vez. *Sugestão:* o
operador `in` não é necessário e vai dificultar a solução.

| Argumentos | Retorno |
| ---------- | ------- |
| `[1,1,4,5,1,5], 5` | `2` |
| `[0,9,2,1,0], 3` | `0` |
| `[0,9,2,1,0], 1` | `1` |

**b) Função `apenasRepetidos`:** recebe uma lista e retorna uma nova lista contendo
apenas os elementos que aparecem **mais de uma vez** na lista recebida. A função também
retorna o **menor** elemento dentre os repetidos, como **último elemento** da lista de
retorno. As duas análises devem ser feitas em uma única estrutura de repetição.
*Sugestão:* use a função da letra a) para verificar se um elemento aparece mais de uma
vez.

| Argumento | Retorno |
| --------- | ------- |
| `[2, 3, 4, 1, 3, 1, 0]` | `[3, 1, 1]` |
| `[0, 9, 2, 1]` | `[]` |
| `[0, 9, 2, 1, 9]` | `[9, 9]` |
| `[]` | `[]` |

````python
def quantidades(lista, numero):
    total = 0
    for elemento in lista:
        if elemento == numero:
            total = total + 1
    return total


def apenasRepetidos(lista):
    resultado = []
    menor = 0
    for elemento in lista:
        if quantidades(lista, elemento) > 1 and quantidades(resultado, elemento) == 0:
            if len(resultado) == 0 or elemento < menor:
                menor = elemento
            resultado.append(elemento)

    if len(resultado) > 0:
        resultado.append(menor)
    return resultado


print(quantidades([1, 1, 4, 5, 1, 5], 5))
print(apenasRepetidos([2, 3, 4, 1, 3, 1, 0]))
print(apenasRepetidos([0, 9, 2, 1]))
print(apenasRepetidos([0, 9, 2, 1, 9]))
print(apenasRepetidos([]))
````

`quantidades(resultado, elemento) == 0` garante que cada elemento repetido entre só uma
vez no resultado. O menor é atualizado no mesmo laço, como a questão pede.
