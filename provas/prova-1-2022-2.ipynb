### Questão 1

Desenvolva um programa que leia um número inteiro maior do que 1 (não é necessário
verificar) e mostre uma figura como nos exemplos abaixo. A entrada determina a altura e
a largura da figura.

*Sugestão:* crie e use uma função `mostraLinha`, que recebe uma quantidade de espaços e
uma quantidade de números e mostra os espaços seguidos dos números em uma única linha.
Essa função não precisa ter retorno.

```
Entrada 2     Entrada 3     Entrada 4
 1              1              1
123            123            123
 1            12345          12345
               123          1234567
                1            12345
                              123
                               1
```

````python
def mostraLinha(espacos, numeros):
    contador = 1
    while contador <= espacos:
        print(' ', end='')
        contador += 1
    contador = 1
    while contador <= numeros:
        print(contador, end='')
        contador += 1
    print()


n = int(input('Entrada: '))

linha = 1
while linha <= n:                 # parte de cima, incluindo a linha do meio
    mostraLinha(n - linha, 2 * linha - 1)
    linha += 1

linha = n - 1
while linha >= 1:                 # parte de baixo
    mostraLinha(n - linha, 2 * linha - 1)
    linha -= 1
````

Monte a tabela para a entrada 4: a linha `l` tem `4 - l` espaços e `2 * l - 1`
números. Com a função `mostraLinha`, o programa principal só precisa calcular esses
dois valores para cada linha.

### Questão 2

Desenvolva um programa com as funções das letras a) e b).

**a) Função `indiceDoElemento`:** recebe uma lista e um número. Se o número estiver na
lista, retorna o seu índice (sempre o da **primeira** ocorrência, mesmo que ele apareça
mais de uma vez). Se não estiver, retorna `-1`. *Sugestão:* o operador `in` não é
necessário e vai dificultar a solução.

**b) Função `mostraQuantidades`:** recebe uma lista e mostra, diretamente na saída,
quantas vezes cada elemento aparece. A função deve percorrer a lista recebida **apenas
uma vez** (nem cópias dela podem ser percorridas) e não tem retorno. *Sugestão:* use a
função da letra a) quando necessário, mas **não** na lista recebida.

```
[1,1,4,5,1,5]  →  O elemento 1 repete-se 3 vezes.
                  O elemento 4 repete-se 1 vez.
                  O elemento 5 repete-se 2 vezes.

[1]            →  O elemento 1 repete-se 1 vez.
```

````python
def indiceDoElemento(lista, numero):
    indice = 0
    while indice < len(lista):
        if lista[indice] == numero:
            return indice
        indice += 1
    return -1


def mostraQuantidades(lista):
    elementos = []      # elementos distintos, na ordem em que aparecem
    quantidades = []    # quantidades[i] corresponde a elementos[i]

    for valor in lista:                       # única passagem pela lista recebida
        posicao = indiceDoElemento(elementos, valor)
        if posicao == -1:
            elementos.append(valor)
            quantidades.append(1)
        else:
            quantidades[posicao] = quantidades[posicao] + 1

    indice = 0
    while indice < len(elementos):
        if quantidades[indice] == 1:
            print('O elemento', elementos[indice], 'repete-se 1 vez.')
        else:
            print('O elemento', elementos[indice], 'repete-se', quantidades[indice], 'vezes.')
        indice += 1


print(indiceDoElemento([4, 7, 9, 7], 7), indiceDoElemento([4, 7], 3))
mostraQuantidades([1, 1, 4, 5, 1, 5])
mostraQuantidades([0, 9, 2, 1, 0])
````

O truque é usar `indiceDoElemento` na lista `elementos` (que vamos montando), e não na
lista recebida. Assim, a lista recebida é percorrida só uma vez.
