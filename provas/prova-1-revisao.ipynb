### Exercício 1

Mostre a figura abaixo, formada por números, conforme a entrada do usuário:

```
entrada = 2      entrada = 4
1                1
123              123
1                12345
                 1234567
                 12345
                 123
                 1
```

````python
n = int(input('Entrada: '))

# Cada linha mostra os números de 1 até um limite ímpar: 1, 3, 5... e depois volta
limite = 1
while limite <= 2 * n - 1:
    numero = 1
    while numero <= limite:
        print(numero, end='')
        numero += 1
    print()
    limite += 2

limite = 2 * n - 3
while limite >= 1:
    numero = 1
    while numero <= limite:
        print(numero, end='')
        numero += 1
    print()
    limite -= 2
````

### Exercício 2

Mostre a figura abaixo, formada por números, conforme a entrada do usuário:

```
entrada = 2      entrada = 4
1                1
121              121
1                12321
                 1234321
                 12321
                 121
                 1
```

````python
n = int(input('Entrada: '))

# A linha "topo" sobe de 1 até topo e desce de volta: topo = 4 -> 1234321
topo = 1
subindo = True
while topo >= 1:
    numero = 1
    while numero <= topo:
        print(numero, end='')
        numero += 1
    numero = topo - 1
    while numero >= 1:
        print(numero, end='')
        numero -= 1
    print()

    if topo == n:
        subindo = False
    if subindo:
        topo += 1
    else:
        topo -= 1
````

Aqui usamos uma variável lógica (`subindo`) para controlar se a figura está crescendo ou
diminuindo. Assim o desenho de cada linha fica escrito uma vez só.

### Exercício 3

Obtenha pela entrada do usuário dois números inteiros que representam os
limites inferior e superior de um intervalo; depois, uma quantidade `n` de números a
serem lidos; e, por fim, os `n` números inteiros.

- O limite superior precisa ser maior que o inferior e a distância entre eles deve ser
  maior do que 2. Caso contrário, mostre `intervalo inválido` e encerre o programa.
- A quantidade `n` deve ser maior do que 0. Caso contrário, mostre `quantidade inválida`
  e encerre o programa.

Ao final, informe os **dois maiores** números entre os que pertencem ao intervalo
(incluindo os limites). Se algum número se repetir, considere-o apenas uma vez. Se nenhum
número satisfizer os critérios, mostre `nenhum número`; se apenas um satisfizer, mostre só
ele.

````python
inferior = int(input('Limite inferior: '))
superior = int(input('Limite superior: '))

if superior - inferior <= 2:
    print('intervalo inválido')
else:
    n = int(input('Quantidade de números: '))
    if n <= 0:
        print('quantidade inválida')
    else:
        encontrados = 0   # quantos números distintos do intervalo já guardamos (0, 1 ou 2)
        maior = 0
        segundo = 0

        lidos = 0
        while lidos < n:
            numero = int(input('Número: '))
            no_intervalo = inferior <= numero <= superior
            repetido = (encontrados >= 1 and numero == maior) or (encontrados == 2 and numero == segundo)

            if no_intervalo and not repetido:
                if encontrados == 0:
                    maior = numero
                    encontrados = 1
                elif numero > maior:
                    segundo = maior
                    maior = numero
                    encontrados = 2
                elif encontrados == 1 or numero > segundo:
                    segundo = numero
                    encontrados = 2
            lidos += 1

        if encontrados == 0:
            print('nenhum número')
        elif encontrados == 1:
            print(maior)
        else:
            print(maior, segundo)
````

### Exercício 4

Leia números inteiros até que o usuário digite `-555`. Mostre o somatório dos
números que **não** terminam com o dígito 3, nem com 5, nem com 7.

````python
numero = int(input('Número (-555 para sair): '))
soma = 0

while numero != -555:
    ultimo_digito = abs(numero) % 10
    if ultimo_digito != 3 and ultimo_digito != 5 and ultimo_digito != 7:
        soma += numero
    numero = int(input('Número (-555 para sair): '))

print(f'Somatório: {soma}')
````

Usamos `abs(numero) % 10` por causa dos números negativos: em Python, `-17 % 10` dá `3`
(e não `7`), porque o resto acompanha o sinal do divisor. Com `abs`, `-17` termina em `7`,
como esperado.
