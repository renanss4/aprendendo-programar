````python
class Produto:
    def __init__(self, ean='', fabricante='', custoFabrica=0):
        self._ean = ean                       # texto
        self._fabricante = fabricante         # texto
        self.__custoFabrica = custoFabrica    # ponto flutuante

    def calcularCusto(self):
        return self.__custoFabrica * 1.12

    def getEan(self):           # retorna um texto
        return self._ean

    def getFabricante(self):    # retorna um texto
        return self._fabricante
````

### Questão 1

Desenvolva uma **função** que receba uma matriz quadrada (caso não seja, retorne uma
lista vazia) e **retorne uma lista com os índices das diagonais acima da secundária
que têm todos os valores diferentes entre si**. Verifique apenas as diagonais com mais
de um elemento. A diagonal mais próxima da secundária tem índice 1, e os índices das
demais aumentam de 1 em 1. Se nenhuma diagonal atender ao critério, retorne uma lista
vazia.

```
Parâmetro        Retorno        Parâmetro         Retorno
1  1 -4          [1]            1 2 3 4 6         [2, 3]
2 -1 -2                         1 2 4 5 7
3 -2  1                         1 2 5 7 8
                                0 5 7 1 9
1 2 4            []             1 2 5 7 8
2 1 2
3 1 1                           8 -3 1            []
                                4 -4 2
```

````python
def eh_quadrada(matriz):
    for linha in matriz:
        if len(linha) != len(matriz):
            return False
    return True


def diagonais_diferentes_acima_secundaria(matriz):
    if not eh_quadrada(matriz):
        return []

    n = len(matriz)
    resultado = []
    k = 1
    while k <= n - 2:
        # elementos da diagonal k: linha + coluna == n - 1 - k
        vistos = []
        todos_diferentes = True
        linha = 0
        while linha <= n - 1 - k:
            valor = matriz[linha][n - 1 - k - linha]
            if valor in vistos:
                todos_diferentes = False
            vistos.append(valor)
            linha += 1
        if todos_diferentes:
            resultado.append(k)
        k += 1
    return resultado


print(diagonais_diferentes_acima_secundaria([[1, 1, -4], [2, -1, -2], [3, -2, 1]]))
print(diagonais_diferentes_acima_secundaria([[1, 2, 4], [2, 1, 2], [3, 1, 1]]))
print(diagonais_diferentes_acima_secundaria([
    [1, 2, 3, 4, 6], [1, 2, 4, 5, 7], [1, 2, 5, 7, 8], [0, 5, 7, 1, 9], [1, 2, 5, 7, 8]]))
print(diagonais_diferentes_acima_secundaria([[8, -3, 1], [4, -4, 2]]))
````

### Questão 2

**2.1** Construa a classe `ProdutoImportado`, subclasse da classe `Produto` (código
fornecido na célula de preparação). Além dos atributos herdados, ela tem os atributos
**protegidos** `pais` e `taxa`.

- a) **Construtor:** tem parâmetros para inicializar todos os atributos da classe.
  Todos são opcionais, com os valores padrão `'Japão'` para `pais` (texto) e 13% para
  `taxa` (ponto flutuante). É obrigatório executar o construtor da superclasse (com
  `super`) para inicializar os atributos herdados.
- b) Sobrescreva o método `calcularCusto` para retornar o custo do produto importado:
  aplique o percentual da taxa de importação ao custo normal do produto (calculado pelo
  mesmo método da superclasse). Exemplo: custo 100, taxa 15% → retorno 115.

```python
class Produto:
    def __init__(self, ean='', fabricante='', custoFabrica=0):
        self._ean = ean                       # texto
        self._fabricante = fabricante         # texto
        self.__custoFabrica = custoFabrica    # ponto flutuante

    def calcularCusto(self):
        return self.__custoFabrica * 1.12

    def getEan(self):           # retorna um texto
        return self._ean

    def getFabricante(self):    # retorna um texto
        return self._fabricante
```

````python
class ProdutoImportado(Produto):
    def __init__(self, ean='', fabricante='', custoFabrica=0, pais='Japão', taxa=13.0):
        super().__init__(ean, fabricante, custoFabrica)
        self._pais = pais
        self._taxa = taxa      # percentual: 13.0 significa 13%

    def calcularCusto(self):
        custo_normal = super().calcularCusto()
        return custo_normal + custo_normal * self._taxa / 100


nacional = Produto('789', 'Acme', 100)
importado = ProdutoImportado('123', 'Sony', 100, taxa=15.0)
print(f'{nacional.calcularCusto():.2f} {importado.calcularCusto():.2f}')
````

Guardamos a taxa como percentual (`13.0`), que é como o enunciado a apresenta. Também
seria válido guardar `0.13`, desde que o cálculo seja ajustado.

- b) **Método `produtosFabricante`:** recebe o nome de um fabricante e retorna uma lista
  com todos os produtos (objetos) desse fabricante no cadastro. Se não houver nenhum,
  retorna uma lista vazia.

class ECommerce:

````python
class ECommerce:
    def __init__(self):
        self.__cadastro = {}

    def adicionaProduto(self, prod):
        self.__cadastro[prod.getEan()] = prod

    def custoProduto(self, ean):
        if ean not in self.__cadastro:
            return False
        return self.__cadastro[ean].calcularCusto()

    def produtosFabricante(self, fabricante):
        produtos = []
        for produto in self.__cadastro.values():
            if produto.getFabricante() == fabricante:
                produtos.append(produto)
        return produtos


loja = ECommerce()
loja.adicionaProduto(Produto('789', 'Acme', 100))
loja.adicionaProduto(ProdutoImportado('123', 'Sony', 100))
loja.adicionaProduto(ProdutoImportado('456', 'Sony', 50, 'China', 20.0))
print(f"{loja.custoProduto('123'):.2f}", loja.custoProduto('000'))
print(len(loja.produtosFabricante('Sony')), loja.produtosFabricante('Nokia'))
````

`custoProduto` chama `calcularCusto` sem saber se o objeto é um `Produto` ou um
`ProdutoImportado`; cada classe responde com a sua versão do método. Isso é
polimorfismo.
