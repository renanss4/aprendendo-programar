### Questão 1

Desenvolva uma **função** que receba um número inteiro maior do que 1 (não é necessário
verificar) e **retorne uma matriz** com uma figura como nos exemplos abaixo. O
parâmetro determina a altura e a largura da figura e, portanto, as dimensões da matriz.
Nas posições em que não houver o caractere `'*'`, coloque o caractere `' '` (espaço).

```
Parâmetro 2     Parâmetro 3     Parâmetro 5
 **               ***               *****
**               ***               *****
                ***               *****
                                 *****
                                *****
```

````python
def figura(n):
    matriz = []
    linha = 0
    while linha < n:
        nova_linha = []
        coluna = 0
        while coluna < 2 * n - 1:
            # na linha l, os asteriscos começam na coluna n - 1 - l
            inicio = n - 1 - linha
            if inicio <= coluna < inicio + n:
                nova_linha.append('*')
            else:
                nova_linha.append(' ')
            coluna += 1
        matriz.append(nova_linha)
        linha += 1
    return matriz


for linha in figura(5):
    for caractere in linha:
        print(caractere, end='')
    print()
````

A matriz tem `n` linhas e `2n - 1` colunas. Na linha `l` (começando em 0), há
`n - 1 - l` espaços, depois `n` asteriscos e o restante são espaços.

### Questão 2

Desenvolva uma **função** que receba uma matriz quadrada (caso não seja, retorne uma
lista vazia) e **retorne uma lista com os índices das diagonais abaixo da secundária
que têm todos os valores iguais**. Verifique apenas as diagonais com mais de um
elemento. A diagonal mais próxima da secundária tem índice 1, e os índices das demais
aumentam de 1 em 1. Se nenhuma diagonal tiver todos os elementos iguais, retorne uma
lista vazia.

```
Parâmetro        Retorno        Parâmetro         Retorno
1  1 -4          [1]             1  2  3  4       [1, 2]
2 -1 -2                         -1  3  4  5
3 -2  1                          1  2  5  7
                                 0  5  7 -1
1 1 4            []              8 -3  1          []
2 1 2                            4 -4  2
3 1 1
```

````python
def eh_quadrada(matriz):
    for linha in matriz:
        if len(linha) != len(matriz):
            return False
    return True


def diagonais_iguais_abaixo_secundaria(matriz):
    if not eh_quadrada(matriz):
        return []

    n = len(matriz)
    resultado = []
    k = 1
    while k <= n - 2:     # a diagonal k tem n - k elementos; queremos mais de 1
        # elementos da diagonal k: linha + coluna == n - 1 + k
        linha = k
        primeiro = matriz[linha][n - 1 + k - linha]
        todos_iguais = True
        while linha < n:
            if matriz[linha][n - 1 + k - linha] != primeiro:
                todos_iguais = False
            linha += 1
        if todos_iguais:
            resultado.append(k)
        k += 1
    return resultado


print(diagonais_iguais_abaixo_secundaria([[1, 1, -4], [2, -1, -2], [3, -2, 1]]))
print(diagonais_iguais_abaixo_secundaria([[1, 1, 4], [2, 1, 2], [3, 1, 1]]))
print(diagonais_iguais_abaixo_secundaria([[1, 2, 3, 4], [-1, 3, 4, 5], [1, 2, 5, 7], [0, 5, 7, -1]]))
print(diagonais_iguais_abaixo_secundaria([[8, -3, 1], [4, -4, 2]]))
````

Na diagonal secundária, `linha + coluna == n - 1`. A diagonal `k` abaixo dela tem
`linha + coluna == n - 1 + k`, começando na linha `k`. Por isso, para cada linha,
a coluna é `n - 1 + k - linha`.

### Questão 3

**3.1** Construa a classe `PlayerCS`, subclasse da classe `Player` (código fornecido na
célula de preparação). Além dos atributos herdados, `PlayerCS` tem os atributos
**privados** `timeFavorito` e `armaFavorita`.

- a) **Construtor:** tem parâmetros para inicializar todos os atributos da classe.
  Todos são opcionais, com os valores padrão `'contra-terrorista'` para `timeFavorito`
  e `'rifle'` para `armaFavorita`. É obrigatório executar o construtor da superclasse
  (com `super`) para inicializar os atributos herdados.
- b) Sobrescreva o método `__str__` para retornar, em um único texto: id, créditos,
  nome do jogo (sempre CS), time favorito e arma favorita. Formato:
  `'player id: 123, créditos: 87, jogo: CS, time favorito: contra-terrorista, arma favorita: sniper'`.

```python
class Player:
    def __init__(self, pl_id=0, cr=0):
        self.__playerID = pl_id
        self.__creditos = cr

    def __str__(self):
        return f'player id: {self.__playerID}, créditos: {self.__creditos}'

    def getPlayerID(self):
        return self.__playerID
```

````python
class PlayerCS(Player):
    def __init__(self, pl_id=0, cr=0, timeFavorito='contra-terrorista', armaFavorita='rifle'):
        super().__init__(pl_id, cr)
        self.__timeFavorito = timeFavorito
        self.__armaFavorita = armaFavorita

    def __str__(self):
        return (super().__str__() + ', jogo: CS'
                + f', time favorito: {self.__timeFavorito}'
                + f', arma favorita: {self.__armaFavorita}')


print(PlayerCS(123, 87, armaFavorita='sniper'))
print(PlayerCS())
````

Os atributos `__playerID` e `__creditos` são privados de `Player`, então `PlayerCS` não
consegue acessá-los diretamente. A saída é aproveitar o `__str__` da superclasse com
`super().__str__()`.

**3.2** Construa uma classe `CadastroPlayers` com apenas um atributo, `cadastro`, que
deve ser **privado** e guardar um dicionário.

- a) **Construtor:** não recebe nenhuma informação; apenas cria o atributo com um
  dicionário vazio.
- b) **Método `adicionaPlayer`:** recebe um objeto do tipo `Player` ou `PlayerCS` (não é
  necessário verificar o tipo) e o adiciona ao cadastro. Se o objeto já estiver no
  dicionário, o método retorna `False` e o objeto não é inserido. Caso contrário,
  adicione-o usando o *player id* como chave e o próprio objeto como valor. O player id
  é único para cada jogador.

````python
class CadastroPlayers:
    def __init__(self):
        self.__cadastro = {}

    def adicionaPlayer(self, player):
        if player.getPlayerID() in self.__cadastro:
            return False
        self.__cadastro[player.getPlayerID()] = player
        return True


cadastro = CadastroPlayers()
print(cadastro.adicionaPlayer(Player(1, 50)))
print(cadastro.adicionaPlayer(PlayerCS(2, 87)))
print(cadastro.adicionaPlayer(PlayerCS(1, 10)))   # id repetido
````
