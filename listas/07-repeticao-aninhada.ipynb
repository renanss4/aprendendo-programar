## Teoria

## Um laço dentro do outro

Um laço **aninhado** é um `while` dentro de outro. Para cada volta do laço de fora, o laço
de dentro executa **todas** as suas voltas. É como um relógio: para cada hora, o ponteiro
dos minutos dá uma volta completa.

````python
hora = 1
while hora <= 2:
    minuto = 0
    while minuto < 60:
        if minuto % 15 == 0:
            print(f'{hora}h{minuto:02d}')
        minuto += 1
    hora += 1
````

> Repare que `minuto = 0` fica **dentro** do laço de fora. Se ficasse antes, o laço de
> dentro só funcionaria na primeira hora.

## Mostrando na mesma linha

Por padrão, o `print` pula de linha no final. Com `end=''`, ele não pula, e o próximo
`print` continua na mesma linha. Um `print()` vazio só pula a linha:

````python
print('a', end='')
print('b', end='')
print('c')
print('nova linha')
````

## Desenhando figuras

Para desenhar figuras, pense em **linhas** (laço de fora) e **colunas** (laço de dentro).
O segredo é descobrir quantos caracteres cada linha tem em função do número da linha:

````python
altura = 4

linha = 1
while linha <= altura:
    coluna = 1
    while coluna <= linha:    # a linha 1 tem 1 asterisco, a linha 2 tem 2...
        print('*', end='')
        coluna += 1
    print()                   # terminou a linha: pula para a próxima
    linha += 1
````

Uma dica para figuras mais complicadas: monte uma tabela com o número da linha e a
quantidade de cada caractere. Por exemplo, para um triângulo centralizado de altura 3:

| Linha | Espaços | Asteriscos |
| ----- | ------- | ---------- |
| 1 | 2 | 1 |
| 2 | 1 | 3 |
| 3 | 0 | 5 |

Daí sai a regra: espaços = `altura - linha` e asteriscos = `2 * linha - 1`.

## Erros comuns

- Inicializar a variável do laço de dentro fora do laço de fora.
- Esquecer o `print()` que pula de linha ao final de cada linha da figura.
- Confundir a variável de um laço com a do outro (dê nomes claros: `linha`, `coluna`).

### Exercício 1

Obtenha dois números inteiros positivos (início e fim) pela entrada do usuário e mostre
as tabuadas de todos os números do intervalo, incluindo início e fim. Exemplo com as
entradas 2 e 6:

```
Tabuada do 2:
2 x 1 = 2
2 x 2 = 4
...
2 x 10 = 20
Tabuada do 3:
3 x 1 = 3
...
Tabuada do 6:
...
```

````python
inicio = int(input('Início: '))
fim = int(input('Fim: '))

numero = inicio
while numero <= fim:
    print(f'Tabuada do {numero}:')
    multiplicador = 1
    while multiplicador <= 10:
        print(f'{numero} x {multiplicador} = {numero * multiplicador}')
        multiplicador += 1
    numero += 1
````

### Exercício 2

Obtenha a quantidade de números que serão lidos (`qt`). Em seguida, leia essa quantidade
de números e, ao final, mostre o somatório dos fatoriais dos números lidos.

````python
qt = int(input('Quantidade de números: '))

soma = 0
lidos = 0
while lidos < qt:
    numero = int(input('Número: '))

    fatorial = 1
    fator = 1
    while fator <= numero:
        fatorial *= fator
        fator += 1

    soma += fatorial
    lidos += 1

print(f'Somatório dos fatoriais: {soma}')
````

O laço de dentro calcula o fatorial de cada número lido. Com as entradas 3, 4 e 0, o
resultado é $3! + 4! + 0! = 6 + 24 + 1 = 31$. Na próxima aula você vai aprender a colocar
esse cálculo em uma **função**, deixando o código mais organizado.

### Exercício 3

Obtenha uma quantidade de linhas (um inteiro maior do que 0) e mostre um triângulo
formado pelo caractere `*` com essa quantidade de linhas. Exemplo para 4 linhas:

```
*
**
***
****
```

````python
linhas = int(input('Quantidade de linhas: '))

linha = 1
while linha <= linhas:
    coluna = 1
    while coluna <= linha:
        print('*', end='')
        coluna += 1
    print()
    linha += 1
````

### Exercício 4

Mesmo enunciado do exercício anterior, porém com o padrão abaixo (exemplo para a
entrada 4):

```
*
**
***
****
***
**
*
```

````python
linhas = int(input('Quantidade de linhas: '))

# Parte de cima: 1 até linhas asteriscos
linha = 1
while linha <= linhas:
    coluna = 1
    while coluna <= linha:
        print('*', end='')
        coluna += 1
    print()
    linha += 1

# Parte de baixo: linhas - 1 até 1 asterisco
linha = linhas - 1
while linha >= 1:
    coluna = 1
    while coluna <= linha:
        print('*', end='')
        coluna += 1
    print()
    linha -= 1
````
