## Teoria

## Diagonais de uma matriz quadrada

Em uma matriz **quadrada** (mesma quantidade de linhas e colunas, `n x n`):

- a **diagonal principal** tem os elementos em que `linha == coluna`;
- a **diagonal secundária** tem os elementos em que `linha + coluna == n - 1`;
- as diagonais **acima** da principal têm `coluna - linha = k`, com `k = 1, 2, ...`;
- as diagonais **abaixo** da principal têm `linha - coluna = k`.

```
principal      secundária
[1] 2  3        1  2 [3]
 4 [5] 6        4 [5] 6
 7  8 [9]      [7] 8  9
```

````python
m = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]
n = len(m)

i = 0
while i < n:
    print('principal:', m[i][i], ' secundária:', m[i][n - 1 - i])
    i += 1
````

## Erros comuns

- Inverter linha e coluna: é `m[linha][coluna]`.
- Usar `len(matriz)` como quantidade de colunas: isso é a quantidade de **linhas**.
  Colunas são `len(matriz[0])`.
- Criar a matriz com `[[0] * colunas] * linhas` (as linhas ficam ligadas).
- Esquecer de reiniciar o índice da coluna a cada nova linha.

### Exercício 1

Crie uma função que recebe uma matriz quadrada e retorna a média dos elementos da
diagonal principal.

````python
def media_diagonal_principal(matriz):
    soma = 0
    i = 0
    while i < len(matriz):
        soma += matriz[i][i]
        i += 1
    return soma / len(matriz)

print(media_diagonal_principal([[1, 2, 3], [3, 2, 1], [2, 3, 6]]))
````

### Exercício 2

Crie uma função que recebe uma matriz quadrada e retorna `True` se todos os elementos da
diagonal secundária forem iguais e `False` caso contrário.

````python
def secundaria_constante(matriz):
    n = len(matriz)
    primeiro = matriz[0][n - 1]
    i = 1
    while i < n:
        if matriz[i][n - 1 - i] != primeiro:
            return False
        i += 1
    return True

print(secundaria_constante([[1, 2, 4], [3, 4, 1], [4, 0, 9]]))
print(secundaria_constante([[1, 2], [3, 4]]))
````

### Exercício 3

Crie uma função que recebe uma matriz quadrada e mostra a matriz multiplicada pelos
elementos da sua diagonal principal: multiplique todos os elementos de cada linha pelo
elemento da diagonal principal daquela linha.

````python
def mostrar_multiplicada_pela_principal(matriz):
    linha = 0
    while linha < len(matriz):
        fator = matriz[linha][linha]
        for elemento in matriz[linha]:
            print(elemento * fator, end=' ')
        print()
        linha += 1

mostrar_multiplicada_pela_principal([[2, 2, 3], [4, 3, 4], [10, 4, 8]])
````

### Exercício 4

Crie uma função que recebe uma matriz quadrada e verifica se ela é um **quadrado
mágico**: uma matriz de inteiros em que a soma de cada linha, a soma de cada coluna e as
somas das diagonais principal e secundária são todas iguais.

Exemplo: `[[8, 0, 7], [4, 5, 6], [3, 10, 2]]`

````python
def eh_quadrado_magico(matriz):
    n = len(matriz)
    alvo = sum(matriz[0])   # todas as somas precisam ser iguais à da primeira linha

    soma_principal = 0
    soma_secundaria = 0
    i = 0
    while i < n:
        soma_linha = 0
        soma_coluna = 0
        j = 0
        while j < n:
            soma_linha += matriz[i][j]
            soma_coluna += matriz[j][i]
            j += 1
        if soma_linha != alvo or soma_coluna != alvo:
            return False
        soma_principal += matriz[i][i]
        soma_secundaria += matriz[i][n - 1 - i]
        i += 1

    return soma_principal == alvo and soma_secundaria == alvo

print(eh_quadrado_magico([[8, 0, 7], [4, 5, 6], [3, 10, 2]]))
print(eh_quadrado_magico([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))
````

No exemplo do enunciado, todas as linhas, colunas e as duas diagonais somam 15. Não basta
comparar só as diagonais: é preciso verificar também cada linha e cada coluna.

### Exercício 5

Crie uma função que recebe uma matriz quadrada e informa quais diagonais, acima e abaixo
da diagonal principal, possuem o valor 0. Não repita diagonais na saída.

````python
def diagonais_com_zero(matriz):
    n = len(matriz)
    k = 1
    while k < n:
        tem_zero_acima = False
        tem_zero_abaixo = False
        i = 0
        while i + k < n:
            if matriz[i][i + k] == 0:    # diagonal k acima da principal
                tem_zero_acima = True
            if matriz[i + k][i] == 0:    # diagonal k abaixo da principal
                tem_zero_abaixo = True
            i += 1
        if tem_zero_acima:
            print(f'A diagonal {k} acima da principal tem o valor 0')
        if tem_zero_abaixo:
            print(f'A diagonal {k} abaixo da principal tem o valor 0')
        k += 1

diagonais_com_zero([
    [1, 0, 2, 0],
    [5, 1, 0, 3],
    [0, 6, 1, 0],
    [7, 0, 8, 1],
])
````

A diagonal `k` acima da principal tem os elementos `matriz[i][i + k]`; a diagonal `k`
abaixo tem `matriz[i + k][i]`. Usar uma variável lógica para cada diagonal garante que
cada uma seja informada só uma vez, mesmo que tenha vários zeros.
