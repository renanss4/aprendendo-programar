## Teoria

## Variáveis

Uma **variável** é um nome que guarda um valor. Criamos uma variável com o sinal `=`
(que em programação significa **atribuição**, e não igualdade):

````python
idade = 25
nome = 'Maria'
salario = 1500.50

print(nome)
print(idade)
print(salario)
````

Cada valor tem um **tipo**. Os mais comuns são:

| Tipo | Significado | Exemplo |
| ---- | ----------- | ------- |
| `int` | número inteiro | `25` |
| `float` | número real | `1500.50` |
| `str` | texto (*string*) | `'Maria'` |
| `bool` | lógico | `True` ou `False` |

Use `type` para descobrir o tipo de um valor:

````python
print(type(idade))
print(type(salario))
print(type(nome))
````

**Boas práticas para nomes de variáveis:** use letras minúsculas, separe palavras com `_`
(`preco_total`) e escolha nomes que digam o que a variável guarda (`area`, e não `x`).

## Entrada de dados com `input`

A função `input` mostra uma mensagem, espera o usuário digitar algo e devolve o que foi
digitado **sempre como texto** (`str`). Para fazer contas, converta o texto com `int` ou
`float`:

````python
nome = input('Digite seu nome: ')
idade = int(input('Digite sua idade: '))
altura = float(input('Digite sua altura em metros: '))

print(nome, 'tem', idade, 'anos e', altura, 'm de altura')
````

## Saída formatada com f-strings

Uma **f-string** é um texto com `f` antes das aspas. Dentro dela, tudo o que estiver entre
`{}` é substituído pelo valor da expressão. Depois de `:` você pode definir o formato,
por exemplo `:.2f` para mostrar duas casas decimais:

````python
preco = 7.5
quantidade = 3
total = preco * quantidade

print(f'Total a pagar: R$ {total:.2f}')
print(f'{quantidade} itens de R$ {preco:.2f}')
````

## Erros comuns

- Esquecer de converter a entrada: `input('Idade: ') + 1` dá erro, porque não dá para
  somar texto com número. Use `int(input('Idade: ')) + 1`.
- Usar `int` para valores com casas decimais: `int('1.5')` dá erro. Use `float`.
- Esquecer o `f` da f-string: `'Total: {total}'` mostra literalmente `{total}`.

### Exercício 1

Calcule e mostre a hipotenusa de um triângulo retângulo, obtendo os valores dos catetos
pela entrada do usuário.

Para obter os valores, mostre a mensagem `Informe o valor do cateto N:` (onde `N` indica
qual cateto deve ser informado). Para mostrar o resultado, use a mensagem
`A hipotenusa é: R` (onde `R` é o resultado do cálculo).

$$hip = \sqrt{cat_1^2 + cat_2^2}$$

````python
cateto1 = float(input('Informe o valor do cateto 1: '))
cateto2 = float(input('Informe o valor do cateto 2: '))

hipotenusa = (cateto1 ** 2 + cateto2 ** 2) ** (1 / 2)
print(f'A hipotenusa é: {hipotenusa}')
````

### Exercício 2

Calcule e mostre a área de um triângulo, obtendo os valores da base ($b$) e da altura ($h$)
pela entrada do usuário (os valores podem ser números reais). Mostre mensagens para
pedir as entradas e para a saída.

$$A_{triângulo} = \frac{b \times h}{2}$$

````python
base = float(input('Informe a base do triângulo: '))
altura = float(input('Informe a altura do triângulo: '))

area = base * altura / 2
print(f'A área do triângulo é: {area}')
````

### Exercício 3

Calcule e mostre a área de um trapézio, obtendo os valores da base maior ($B$), da base
menor ($b$) e da altura ($h$) pela entrada do usuário. Mostre mensagens para pedir as
entradas e para a saída.

$$A_{trapézio} = \frac{h \times (b + B)}{2}$$

````python
base_maior = float(input('Informe a base maior: '))
base_menor = float(input('Informe a base menor: '))
altura = float(input('Informe a altura: '))

area = altura * (base_menor + base_maior) / 2
print(f'A área do trapézio é: {area}')
````

### Exercício 4

Seja $f(x) = x^a$ uma função e $f'(x) = a \times x^{(a-1)}$ a sua derivada. Calcule $f'(x)$
com valores de $a$ e $x$ informados pelo usuário e mostre o resultado.

````python
a = float(input('Informe o valor de a: '))
x = float(input('Informe o valor de x: '))

derivada = a * x ** (a - 1)
print(f"f'(x) = {derivada}")
````

### Exercício 7

Obtenha uma quantidade de dias, horas, minutos e segundos pela entrada do usuário.
Calcule e mostre o total em segundos.

```
Entrada: dias: 2; horas: 4; minutos: 40; segundos: 20
Saída: Total em segundos: 189620s

Entrada: dias: 0; horas: 1; minutos: 0; segundos: 20
Saída: Total em segundos: 3620s
```

````python
dias = int(input('dias: '))
horas = int(input('horas: '))
minutos = int(input('minutos: '))
segundos = int(input('segundos: '))

total = dias * 24 * 60 * 60 + horas * 60 * 60 + minutos * 60 + segundos
print(f'Total em segundos: {total}s')
````

### Exercício 8

Calcule o aumento de um salário. O programa deve pedir o valor do salário e a
porcentagem do aumento, e mostrar o valor do aumento e o novo salário (com apenas duas
casas decimais).

```
Qual o valor do salário atual? R$ 1900
Qual o percentual de aumento? 5
O salário com aumento de 5.0% é de R$1995.00

Qual o valor do salário atual? R$ 7450.75
Qual o percentual de aumento? 13
O salário com aumento de 13.0% é de R$8419.35
```

````python
salario = float(input('Qual o valor do salário atual? R$ '))
percentual = float(input('Qual o percentual de aumento? '))

aumento = salario * percentual / 100
novo_salario = salario + aumento

print(f'O valor do aumento é de R${aumento:.2f}')
print(f'O salário com aumento de {percentual}% é de R${novo_salario:.2f}')
````

### Exercício 9

Peça o preço de um produto e o percentual de desconto. Mostre o valor do desconto e o
preço a pagar (com apenas duas casas decimais).

````python
preco = float(input('Qual o preço do produto? R$ '))
percentual = float(input('Qual o percentual de desconto? '))

desconto = preco * percentual / 100
preco_final = preco - desconto

print(f'Desconto: R${desconto:.2f}')
print(f'Preço a pagar: R${preco_final:.2f}')
````

### Exercício 10

Calcule e mostre o tempo de uma viagem de carro. Pergunte a distância a percorrer (em km)
e a velocidade média esperada (em km/h). Mostre o tempo de viagem (em horas) com apenas

### Exercício 11

Converta uma temperatura informada pelo usuário em Celsius para Fahrenheit e mostre o
resultado:

$$F = \frac{9 \times C}{5} + 32$$

````python
celsius = float(input('Temperatura em Celsius: '))

fahrenheit = 9 * celsius / 5 + 32
print(f'{celsius}°C = {fahrenheit}°F')
````

### Exercício 12

Peça a quantidade de km percorridos e a quantidade de dias pelos quais um carro foi
alugado. Calcule e mostre o preço a pagar, sabendo que o aluguel custa R\$ 60 por dia e
R\$ 0,15 por km rodado.

````python
km = float(input('Quantidade de km percorridos: '))
dias = int(input('Quantidade de dias de aluguel: '))

preco = dias * 60 + km * 0.15
print(f'Preço a pagar: R${preco:.2f}')
````

### Exercício 13

Calcule a redução do tempo de vida de um fumante. Pergunte a quantidade de cigarros
fumados por dia e há quantos anos a pessoa fuma. Considere que um fumante perde
10 minutos de vida a cada cigarro. Calcule e mostre a quantidade de dias de vida perdidos.

````python
cigarros_por_dia = int(input('Quantos cigarros você fuma por dia? '))
anos = int(input('Há quantos anos você fuma? '))

total_cigarros = cigarros_por_dia * anos * 365
minutos_perdidos = total_cigarros * 10
dias_perdidos = minutos_perdidos / (60 * 24)

print(f'Você já perdeu aproximadamente {dias_perdidos:.1f} dias de vida.')
````
