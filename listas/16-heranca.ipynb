## Herança: "é um"

Muitas vezes, classes diferentes têm partes em comum. Um estudante e um professor são
ambos **pessoas**: têm nome, telefone, e-mail. Com **herança**, escrevemos o que é comum
uma vez só, em uma **superclasse** (`Pessoa`), e criamos **subclasses** (`Estudante`,
`Professor`) que herdam tudo isso e acrescentam o que é específico.

Uma boa forma de testar se a herança faz sentido é a frase "**é um**": um estudante *é
uma* pessoa, então `Estudante` pode herdar de `Pessoa`.

```python
class Animal:
    def __init__(self, nome):
        self._nome = nome     # protegido: as subclasses podem usar

    def apresentar(self):
        return f'Eu sou {self._nome}'

    def emitir_som(self):
        return '...'


class Cachorro(Animal):       # Cachorro herda de Animal
    def emitir_som(self):     # sobrescreve o método da superclasse
        return 'Au au!'


class Gato(Animal):
    def emitir_som(self):
        return 'Miau!'


rex = Cachorro('Rex')
print(rex.apresentar())     # método herdado de Animal
print(rex.emitir_som())     # método sobrescrito em Cachorro
```

## Reaproveitando o construtor com `super()`

Quando a subclasse tem atributos a mais, ela define o próprio construtor e chama o
construtor da superclasse com `super().__init__(...)`, para não repetir código:

```python
class Veiculo:
    def __init__(self, marca, modelo):
        self._marca = marca
        self._modelo = modelo

    def descricao(self):
        return f'{self._marca} {self._modelo}'


class Carro(Veiculo):
    def __init__(self, marca, modelo, portas=4):
        super().__init__(marca, modelo)   # inicializa marca e modelo
        self._portas = portas

    def descricao(self):
        # aproveita a descrição da superclasse e acrescenta informações
        return super().descricao() + f' ({self._portas} portas)'


print(Veiculo('Honda', 'Biz').descricao())
print(Carro('Fiat', 'Uno', 2).descricao())
```

> Atributos **privados** (`__atributo`) não ficam acessíveis nem para as subclasses. Use
> atributos **protegidos** (`_atributo`) quando as subclasses precisarem deles, ou
> ofereça métodos de acesso na superclasse.

## Polimorfismo

Como todas as subclasses têm o método `emitir_som`, podemos tratar uma lista de animais
diferentes da mesma forma. Cada objeto responde do seu jeito:

```python
animais = [Cachorro('Rex'), Gato('Mimi'), Animal('Genérico')]

for animal in animais:
    print(animal.apresentar(), '-', animal.emitir_som())
```

Para saber se um objeto é de uma classe (ou de uma subclasse dela), use `isinstance`:
`isinstance(rex, Animal)` é `True`.

## Erros comuns

- Esquecer de chamar `super().__init__(...)` no construtor da subclasse: os atributos da
  superclasse não são criados.
- Tentar acessar um atributo privado (`__atributo`) da superclasse dentro da subclasse.
- Usar herança quando a relação é "tem um" (um carro *tem um* motor, mas não *é um*
  motor). Nesse caso, o certo é guardar um objeto dentro do outro.

### Exercício 1

Implemente uma classe `Pessoa` e duas subclasses, `Estudante` e `Funcionario`. Crie
também as classes `Professor` e `Servidor`, subclasses de `Funcionario`. Os atributos
devem ser privados, e só deve ser possível obter e modificar os seus valores por meio de
métodos. Use nomes apropriados para os atributos e métodos.

- a) Uma pessoa tem nome, endereço, telefone e e-mail. Um estudante tem uma graduação
  (calouro, veterano ou formando; só esses valores são aceitos).
- b) Um funcionário tem um escritório (número da sala), salário e data de admissão. Para
  a data, implemente uma classe `Data` com dia, mês e ano (faça as verificações
  necessárias para que não seja possível representar um dia negativo ou 30 de fevereiro,
  por exemplo) e use-a para a data de admissão.
- c) Um professor tem uma quantidade de horas-aula e de horas de pesquisa. Um servidor tem
  uma carga horária (20 ou 40 horas; só esses valores são possíveis) e uma titulação
  (secretário, chefe de expediente etc.).
- d) Para todas as classes, implemente o construtor e use o construtor da superclasse
  para inicializar os atributos herdados.
- e) Todas as classes têm um método `informar_descricao`, que retorna um texto com o nome
  da classe e o nome da pessoa (por exemplo: `'Funcionário Fulano de Tal'`,
  `'Professor Patinhas'`).
- f) Em uma célula abaixo, teste as classes criando objetos e verificando,
  principalmente, se as validações estão corretas e se o método `informar_descricao` foi
  sobrescrito corretamente. *(Adaptado: no original, os testes ficavam em um arquivo
  `.py`.)*

```python
class Data:
    DIAS_NO_MES = {1: 31, 2: 28, 3: 31, 4: 30, 5: 31, 6: 30,
                   7: 31, 8: 31, 9: 30, 10: 31, 11: 30, 12: 31}

    def __init__(self, dia=1, mes=1, ano=2000):
        self.__dia = 1
        self.__mes = 1
        self.__ano = 2000
        self.alterar(dia, mes, ano)

    @staticmethod
    def bissexto(ano):
        return (ano % 4 == 0 and ano % 100 != 0) or ano % 400 == 0

    def alterar(self, dia, mes, ano):
        if mes < 1 or mes > 12:
            return False
        dias = Data.DIAS_NO_MES[mes]
        if mes == 2 and Data.bissexto(ano):
            dias = 29
        if dia < 1 or dia > dias:
            return False
        self.__dia = dia
        self.__mes = mes
        self.__ano = ano
        return True

    def __str__(self):
        return f'{self.__dia:02d}/{self.__mes:02d}/{self.__ano}'


class Pessoa:
    def __init__(self, nome, endereco='', telefone='', email=''):
        self.__nome = nome
        self.__endereco = endereco
        self.__telefone = telefone
        self.__email = email

    def obter_nome(self):
        return self.__nome

    def alterar_nome(self, nome):
        self.__nome = nome

    def obter_endereco(self):
        return self.__endereco

    def alterar_endereco(self, endereco):
        self.__endereco = endereco

    def obter_telefone(self):
        return self.__telefone

    def alterar_telefone(self, telefone):
        self.__telefone = telefone

    def obter_email(self):
        return self.__email

    def alterar_email(self, email):
        self.__email = email

    def informar_descricao(self):
        return f'Pessoa {self.__nome}'


class Estudante(Pessoa):
    GRADUACOES = ['calouro', 'veterano', 'formando']

    def __init__(self, nome, endereco='', telefone='', email='', graduacao='calouro'):
        super().__init__(nome, endereco, telefone, email)
        self.__graduacao = 'calouro'
        self.alterar_graduacao(graduacao)

    def obter_graduacao(self):
        return self.__graduacao

    def alterar_graduacao(self, graduacao):
        if graduacao not in Estudante.GRADUACOES:
            return False
        self.__graduacao = graduacao
        return True

    def informar_descricao(self):
        return f'Estudante {self.obter_nome()}'


class Funcionario(Pessoa):
    def __init__(self, nome, endereco='', telefone='', email='',
                 sala=0, salario=0.0, data_admissao=None):
        super().__init__(nome, endereco, telefone, email)
        self.__sala = sala
        self.__salario = salario
        if data_admissao is None:
            data_admissao = Data()
        self.__data_admissao = data_admissao

    def obter_sala(self):
        return self.__sala

    def alterar_sala(self, sala):
        self.__sala = sala

    def obter_salario(self):
        return self.__salario

    def alterar_salario(self, salario):
        if salario < 0:
            return False
        self.__salario = salario
        return True

    def obter_data_admissao(self):
        return self.__data_admissao

    def informar_descricao(self):
        return f'Funcionário {self.obter_nome()}'


class Professor(Funcionario):
    def __init__(self, nome, horas_aula=0, horas_pesquisa=0, **dados_funcionario):
        super().__init__(nome, **dados_funcionario)
        self.__horas_aula = horas_aula
        self.__horas_pesquisa = horas_pesquisa

    def obter_horas_aula(self):
        return self.__horas_aula

    def alterar_horas_aula(self, horas):
        self.__horas_aula = horas

    def obter_horas_pesquisa(self):
        return self.__horas_pesquisa

    def alterar_horas_pesquisa(self, horas):
        self.__horas_pesquisa = horas

    def informar_descricao(self):
        return f'Professor {self.obter_nome()}'


class Servidor(Funcionario):
    def __init__(self, nome, carga_horaria=40, titulacao='', **dados_funcionario):
        super().__init__(nome, **dados_funcionario)
        self.__carga_horaria = 40
        self.alterar_carga_horaria(carga_horaria)
        self.__titulacao = titulacao

    def obter_carga_horaria(self):
        return self.__carga_horaria

    def alterar_carga_horaria(self, carga):
        if carga != 20 and carga != 40:
            return False
        self.__carga_horaria = carga
        return True

    def obter_titulacao(self):
        return self.__titulacao

    def alterar_titulacao(self, titulacao):
        self.__titulacao = titulacao

    def informar_descricao(self):
        return f'Servidor {self.obter_nome()}'


# Testes
data = Data(30, 2, 2024)          # inválida: fica com o padrão
print(data)
print(data.alterar(29, 2, 2024))  # 2024 é bissexto: válida
print(data)
print(data.alterar(29, 2, 2023))  # 2023 não é bissexto: inválida

estudante = Estudante('Ana', graduacao='veterano')
print(estudante.alterar_graduacao('mestre'), estudante.obter_graduacao())

servidor = Servidor('Carlos', carga_horaria=30, titulacao='secretário')
print(servidor.obter_carga_horaria(), servidor.alterar_carga_horaria(20), servidor.obter_carga_horaria())

professor = Professor('Patinhas', horas_aula=12, sala=301, data_admissao=Data(1, 3, 2015))
print(professor.obter_sala(), professor.obter_data_admissao())

pessoas = [Pessoa('Fulano'), estudante, Funcionario('Fulano de Tal'), professor, servidor]
for pessoa in pessoas:
    print(pessoa.informar_descricao())
```

Alguns detalhes desta solução:

- Os construtores chamam os métodos `alterar_...` para reaproveitar as validações. Se o
  valor for inválido, o atributo fica com um valor padrão válido.
- `Professor` e `Servidor` recebem `**dados_funcionario`: um jeito de aceitar quaisquer
  parâmetros nomeados (`sala=301`, `salario=...`) e repassá-los ao construtor da
  superclasse sem precisar repetir todos.
- `@staticmethod` marca um método que não usa `self`: `bissexto` só depende do ano.
- O último laço mostra o **polimorfismo**: a mesma chamada `informar_descricao()` produz
  textos diferentes para cada classe.

### Exercício 2

Uma revenda de veículos usados vende **carros**, **motos** e **bicicletas elétricas**. Ela
funciona on-line e o seu foco é produzir **anúncios** dos veículos disponíveis. Cada
anúncio mostra o tipo do veículo, a marca, o modelo, o ano de fabricação, a quilometragem
(exceto para bicicletas) e o valor de venda. Quando uma venda é feita, a revenda guarda o
anúncio e o valor final da venda, já que o valor pode ter mudado na negociação. Quando
um novo veículo chega para venda, a revenda obtém os seus dados, cria o anúncio e o
adiciona à **vitrine**, que mostra todos os anúncios disponíveis no momento.

- a) As principais classes são: anúncio, automóvel (e suas subclasses), revenda e venda.
- b) Identifique os métodos e atributos das classes com base na descrição e nas
  funcionalidades a seguir:
    1. novo anúncio
    2. adicionar anúncio
    3. mostrar anúncios
    4. realizar venda
    5. mostrar vendas
    6. mostrar o modelo do veículo vendido pelo maior valor e o respectivo valor
- c) Desenvolva as classes usando subclasses quando necessário e mantendo os atributos
  privados ou protegidos. Crie os métodos de acesso conforme a necessidade.
- d) Teste a classe `Revenda`, que é a principal, garantindo que as funcionalidades do
  item b) estejam corretas. *Sugestão:* ao concluir cada funcionalidade, teste-a
  imediatamente.

```python
class Automovel:
    def __init__(self, marca, modelo, ano):
        self._marca = marca
        self._modelo = modelo
        self._ano = ano

    def obter_modelo(self):
        return self._modelo

    def tipo(self):
        return 'Automóvel'

    def descricao(self):
        return f'{self.tipo()} {self._marca} {self._modelo} ({self._ano})'


class VeiculoMotorizado(Automovel):
    def __init__(self, marca, modelo, ano, quilometragem):
        super().__init__(marca, modelo, ano)
        self._quilometragem = quilometragem

    def descricao(self):
        return super().descricao() + f', {self._quilometragem} km'


class Carro(VeiculoMotorizado):
    def tipo(self):
        return 'Carro'


class Moto(VeiculoMotorizado):
    def tipo(self):
        return 'Moto'


class BicicletaEletrica(Automovel):
    def tipo(self):
        return 'Bicicleta elétrica'


class Anuncio:
    def __init__(self, automovel, valor):
        self.__automovel = automovel
        self.__valor = valor

    def obter_automovel(self):
        return self.__automovel

    def __str__(self):
        return f'{self.__automovel.descricao()} - R$ {self.__valor:.2f}'


class Venda:
    def __init__(self, anuncio, valor_final):
        self.__anuncio = anuncio
        self.__valor_final = valor_final

    def obter_anuncio(self):
        return self.__anuncio

    def obter_valor_final(self):
        return self.__valor_final

    def __str__(self):
        return f'{self.__anuncio} | vendido por R$ {self.__valor_final:.2f}'


class Revenda:
    def __init__(self):
        self.__vitrine = []
        self.__vendas = []

    def novo_anuncio(self, automovel, valor):
        anuncio = Anuncio(automovel, valor)
        self.adicionar_anuncio(anuncio)
        return anuncio

    def adicionar_anuncio(self, anuncio):
        self.__vitrine.append(anuncio)

    def mostrar_anuncios(self):
        print('Vitrine:')
        for anuncio in self.__vitrine:
            print(' ', anuncio)

    def realizar_venda(self, anuncio, valor_final):
        if anuncio not in self.__vitrine:
            return False
        self.__vitrine.remove(anuncio)
        self.__vendas.append(Venda(anuncio, valor_final))
        return True

    def mostrar_vendas(self):
        print('Vendas:')
        for venda in self.__vendas:
            print(' ', venda)

    def mostrar_venda_mais_cara(self):
        if len(self.__vendas) == 0:
            print('Nenhuma venda realizada.')
            return
        mais_cara = self.__vendas[0]
        for venda in self.__vendas:
            if venda.obter_valor_final() > mais_cara.obter_valor_final():
                mais_cara = venda
        modelo = mais_cara.obter_anuncio().obter_automovel().obter_modelo()
        print(f'Venda mais cara: {modelo} por R$ {mais_cara.obter_valor_final():.2f}')


revenda = Revenda()
gol = revenda.novo_anuncio(Carro('VW', 'Gol', 2015, 80000), 35000)
biz = revenda.novo_anuncio(Moto('Honda', 'Biz', 2020, 12000), 9000)
bike = revenda.novo_anuncio(BicicletaEletrica('Caloi', 'E-Vibe', 2022), 7000)
revenda.mostrar_anuncios()

print(revenda.realizar_venda(gol, 33500), revenda.realizar_venda(bike, 7200))
print(revenda.realizar_venda(gol, 1))     # já foi vendido
revenda.mostrar_anuncios()
revenda.mostrar_vendas()
revenda.mostrar_venda_mais_cara()
```

A quilometragem não se aplica a bicicletas, então criamos a classe intermediária
`VeiculoMotorizado`, que acrescenta esse atributo. `Carro` e `Moto` herdam dela;
`BicicletaEletrica` herda direto de `Automovel`. Cada subclasse só sobrescreve o método
`tipo`, e a `descricao` herdada já usa o tipo certo: polimorfismo em ação.
