## Teoria

## O que é uma lista

Uma **lista** guarda vários valores em uma única variável, em ordem. Ela é escrita entre
colchetes, com os elementos separados por vírgula:

```python
notas = [7.5, 9.0, 6.0, 8.5]
nomes = ['Ana', 'Bruno', 'Carla']
mista = [1, 'texto', 3.14, True]   # uma lista pode misturar tipos
vazia = []

print(notas)
print(len(notas))   # len devolve a quantidade de elementos
```

## Índices

Cada elemento tem uma posição, chamada **índice**, que **começa em 0**. Índices negativos
contam a partir do final:

| Elemento | `7.5` | `9.0` | `6.0` | `8.5` |
| -------- | ----- | ----- | ----- | ----- |
| Índice | 0 | 1 | 2 | 3 |
| Índice negativo | -4 | -3 | -2 | -1 |

```python
notas = [7.5, 9.0, 6.0, 8.5]

print(notas[0])     # primeiro
print(notas[-1])    # último
notas[2] = 10.0     # listas podem ser modificadas
print(notas)
```

> Acessar um índice que não existe (por exemplo, `notas[4]` em uma lista de 4 elementos)
> causa o erro `IndexError`. O último índice é sempre `len(lista) - 1`.

## Percorrendo uma lista

Com `while`, usamos um índice que vai de `0` até `len(lista) - 1`:

```python
notas = [7.5, 9.0, 6.0, 8.5]

indice = 0
while indice < len(notas):
    print(f'Nota {indice}: {notas[indice]}')
    indice += 1
```

O `for` percorre os elementos diretamente, sem precisar de índice. Use-o quando não
precisar saber a posição:

```python
notas = [7.5, 9.0, 6.0, 8.5]

soma = 0
for nota in notas:
    soma += nota
print('Média:', soma / len(notas))
```

Outras operações úteis: `elemento in lista` testa se um elemento está na lista e
`lista1 + lista2` cria uma nova lista com os elementos das duas.

## Textos também têm índices

Um texto (`str`) pode ser percorrido como uma lista de caracteres:

```python
palavra = 'abacaxi'
print(palavra[0], palavra[-1], len(palavra))

for letra in palavra:
    print(letra, end=' ')
```

## Importando funções de módulos

O Python tem muitas funções prontas organizadas em **módulos**. Para usar uma delas,
importe-a com `from módulo import função`:

```python
from math import factorial, sqrt
from random import randint
from statistics import mean

print(factorial(5))
print(sqrt(16))
print(randint(1, 6))    # número aleatório de 1 a 6, como um dado
print(mean([7, 8, 9]))
```

## Erros comuns

- Começar a contar os índices em 1: o primeiro elemento é `lista[0]`.
- Usar `<= len(lista)` no `while`: o último índice válido é `len(lista) - 1`.
- Escrever `lista = lista.append(x)`: o `append` modifica a lista e devolve `None`.
  Escreva apenas `lista.append(x)`.
- Inicializar o "maior elemento" com 0: se todos os números forem negativos, o resultado
  fica errado. Comece com o primeiro elemento da lista.

### Exercício 1

Crie uma função que recebe uma lista de números e retorna o maior número da lista.

```python
def maior_elemento(lista):
    maior = lista[0]
    for numero in lista:
        if numero > maior:
            maior = numero
    return maior

print(maior_elemento([3, 9, 2, 7]))
print(maior_elemento([-5, -2, -8]))
```

### Exercício 2

Crie uma função que recebe uma lista de números e retorna quantos números primos a lista
possui. Crie uma função `eh_primo`, que verifica se um número é primo, e use-a para resolver
o exercício. *(Adaptado: no original, a função era importada de um arquivo `.py`.)*

```python
def eh_primo(numero):
    if numero < 2:
        return False
    divisor = 2
    while divisor * divisor <= numero:
        if numero % divisor == 0:
            return False
        divisor += 1
    return True


def contar_primos(lista):
    quantidade = 0
    for numero in lista:
        if eh_primo(numero):
            quantidade += 1
    return quantidade

print(contar_primos([2, 3, 4, 5, 8, 11, 15]))
```

### Exercício 3

Crie uma função que recebe uma lista com valores de tipos diferentes e mostra os valores
em ordem inversa. Essa função não tem retorno.

```python
def mostrar_invertida(lista):
    indice = len(lista) - 1
    while indice >= 0:
        print(lista[indice])
        indice -= 1

mostrar_invertida([1, 'abc', 3.14, True, 'Olá'])
```

### Exercício 4

Crie uma função que recebe uma lista de números e retorna o somatório dos números pares
da lista.

```python
def soma_pares(lista):
    soma = 0
    for numero in lista:
        if numero % 2 == 0:
            soma += numero
    return soma

print(soma_pares([1, 2, 3, 4, 5, 6]))
```

### Exercício 5

Crie uma função que recebe uma lista de números e um número qualquer, e retorna um valor
lógico indicando se o número está presente na lista.

```python
def contem(lista, procurado):
    for numero in lista:
        if numero == procurado:
            return True
    return False

print(contem([1, 2, 3], 2))
print(contem([1, 2, 3], 7))
```

O Python já tem o operador `in` (`procurado in lista`), que faz exatamente isso. Fazer o
laço "na mão" ajuda a entender o que o `in` faz por baixo dos panos. Repare no `return`
dentro do laço: assim que o número é encontrado, a função termina.

### Exercício 6

Crie uma função que recebe duas listas de números e retorna verdadeiro caso a primeira
lista tenha um somatório maior que o da segunda. Para calcular o somatório, use a função
[`sum`](https://docs.python.org/pt-br/3/library/functions.html#sum), que já vem com o

### Exercício 8

Crie uma função que recebe uma lista de notas (valores reais entre 0 e 10) e retorna uma
lista contendo a **média** e a **moda** das notas. Para os cálculos, use funções do módulo
[`statistics`](https://docs.python.org/pt-br/3/library/statistics.html). Caso a lista
contenha alguma nota inválida, não faça o cálculo e a função não retorna nada.

```python
from statistics import mean, mode

def media_e_moda(notas):
    for nota in notas:
        if nota < 0 or nota > 10:
            return None
    return [mean(notas), mode(notas)]

print(media_e_moda([7.5, 8.0, 7.5, 9.0]))
print(media_e_moda([7.5, 11.0]))
```

A **moda** é o valor que mais se repete. No exemplo, `7.5` aparece duas vezes.

### Exercício 9

O gabarito de uma prova com 10 questões (respostas A, B, C, D ou E) é:
`A, A, C, E, D, B, C, E, B, D`. Crie uma função que recebe as respostas de uma prova e
retorna a quantidade de acertos.

```python
GABARITO = ['A', 'A', 'C', 'E', 'D', 'B', 'C', 'E', 'B', 'D']

def contar_acertos(respostas):
    acertos = 0
    indice = 0
    while indice < len(GABARITO):
        if respostas[indice] == GABARITO[indice]:
            acertos += 1
        indice += 1
    return acertos

print(contar_acertos(['A', 'B', 'C', 'E', 'D', 'B', 'A', 'E', 'B', 'C']))
```

Aqui precisamos do índice para comparar a mesma posição nas duas listas, por isso usamos
`while` em vez de `for`.

### Exercício 10

Um texto em Python pode ser acessado por índices, como uma lista: se
`palavra = 'abacaxi'`, então `palavra[0]` é `'a'` e `palavra[1]` é `'b'`. A função `len`
também funciona com textos: `len(palavra)` é `7`.

Crie uma função que recebe um texto e retorna um valor lógico indicando se ele é um
**palíndromo**: uma palavra ou frase que pode ser lida da mesma forma de trás para
frente. Exemplos: reger, rir, radar, ele, esse, ama, aia.

```python
def eh_palindromo(texto):
    inicio = 0
    fim = len(texto) - 1
    while inicio < fim:
        if texto[inicio] != texto[fim]:
            return False
        inicio += 1
        fim -= 1
    return True

print(eh_palindromo('radar'))
print(eh_palindromo('python'))
```

Comparamos o primeiro caractere com o último, o segundo com o penúltimo, e assim por
diante, até os índices se encontrarem no meio. Em Python também dá para inverter um texto
com `texto[::-1]`, mas fazer à mão treina o uso de índices.

### Exercício 11

Crie uma função que recebe um texto e retorna uma lista com 3 elementos: a quantidade de
**vogais**, de **consoantes** e de **números** presentes no texto.

*Dica:* todo caractere tem uma representação numérica (veja a
[tabela ASCII](https://pt.wikipedia.org/wiki/ASCII)), então é possível compará-los com os
operadores relacionais: `'a' <= 'z'` é `True`, e `'1' > '2'` é `False`.

```python
def contar_caracteres(texto):
    vogais = 0
    consoantes = 0
    numeros = 0
    for caractere in texto.lower():
        if caractere in 'aeiou':
            vogais += 1
        elif 'a' <= caractere <= 'z':
            consoantes += 1
        elif '0' <= caractere <= '9':
            numeros += 1
    return [vogais, consoantes, numeros]

print(contar_caracteres('Python 3 é legal!'))
```

Caracteres acentuados (como `é`) e símbolos (espaço, `!`) não entram em nenhuma das três
categorias. Tratar acentos seria um bom desafio extra!
