## Classes e objetos

Até agora, guardamos dados em variáveis, listas e dicionários, e escrevemos funções
separadas para manipulá-los. A **Programação Orientada a Objetos (POO)** junta os dados e
as operações sobre eles em uma mesma estrutura: o **objeto**.

- Uma **classe** é o molde: descreve quais dados (**atributos**) e quais ações
  (**métodos**) os objetos terão.
- Um **objeto** é algo criado a partir do molde. De uma classe `Carro`, podemos criar
  vários carros diferentes.

```python
class Cachorro:
    def __init__(self, nome, raca):
        self.nome = nome
        self.raca = raca

    def latir(self):
        print(f'{self.nome} diz: Au au!')


rex = Cachorro('Rex', 'vira-lata')
bidu = Cachorro('Bidu', 'schnauzer')

rex.latir()
bidu.latir()
print(bidu.raca)
```

- `class Cachorro:` define a classe. Por convenção, nomes de classes começam com letra
  maiúscula.
- `__init__` é o **construtor**: é chamado automaticamente quando criamos um objeto com
  `Cachorro('Rex', 'vira-lata')`, e inicializa os atributos.
- `self` é o próprio objeto. `self.nome = nome` guarda o nome **dentro** do objeto.
- Todo método recebe `self` como primeiro parâmetro, mas não o informamos na chamada:
  `rex.latir()`.

## Encapsulamento

Às vezes não queremos que um atributo seja alterado livremente. Em uma conta bancária,
o saldo só deve mudar por meio de depósitos e saques válidos. Um atributo que começa com
**dois sublinhados** (`__saldo`) é **privado**: só pode ser usado dentro da classe.

```python
class Conta:
    def __init__(self, titular):
        self.__titular = titular
        self.__saldo = 0

    def depositar(self, valor):
        if valor <= 0:
            return False
        self.__saldo += valor
        return True

    def obter_saldo(self):
        return self.__saldo


conta = Conta('Ana')
conta.depositar(100)
print(conta.depositar(-50))   # depósito inválido: False
print(conta.obter_saldo())
```

Tentar acessar `conta.__saldo` fora da classe causa um erro. A única forma de consultar o
saldo é pelo método `obter_saldo`, e a única forma de alterá-lo é pelos métodos que fazem
as validações. Métodos que só devolvem um atributo são chamados de *getters*.

> Um atributo com **um** sublinhado (`_saldo`) é **protegido**: por convenção, indica que
> ele não deve ser usado fora da classe e das suas subclasses (que você vai ver na
> próxima aula).

## Representando um objeto como texto: `__str__`

O método especial `__str__` define o texto mostrado quando fazemos `print(objeto)`:

```python
class Ponto:
    def __init__(self, x=0, y=0):    # parâmetros com valor padrão
        self.__x = x
        self.__y = y

    def __str__(self):
        return f'({self.__x}, {self.__y})'


print(Ponto(3, 4))
print(Ponto())
```

## Objetos que guardam objetos

Um atributo pode guardar outros objetos, por exemplo, em uma lista ou em um dicionário.
Assim, uma classe pode **gerenciar** várias instâncias de outra:

```python
class Agenda:
    def __init__(self):
        self.__contatos = {}    # nome -> objeto Ponto (só para exemplo)

    def adicionar(self, nome, ponto):
        if nome in self.__contatos:
            return False
        self.__contatos[nome] = ponto
        return True

    def listar(self):
        for nome, ponto in self.__contatos.items():
            print(nome, ponto)


agenda = Agenda()
agenda.adicionar('casa', Ponto(1, 2))
agenda.adicionar('trabalho', Ponto(5, 8))
agenda.listar()
```

## Erros comuns

- Esquecer o `self` como primeiro parâmetro de um método.
- Escrever `nome = nome` no construtor em vez de `self.nome = nome` (o valor não fica
  guardado no objeto).
- Chamar o método sem parênteses: `conta.obter_saldo` é o método; `conta.obter_saldo()`
  é o resultado.
- Tentar acessar um atributo privado de fora da classe.

### Exercício 1

Crie uma classe para representar uma **conta bancária** que permita saques, depósitos,
verificação do saldo, informação se o saldo está negativo e troca da senha do dono da
conta. Atenção ao contexto do problema: não é possível sacar valores negativos nem
valores maiores que o saldo. Depósitos também não podem ser negativos nem de R\$ 0,00.
Teste a classe executando os métodos em, no mínimo, dois objetos diferentes.

```python
class ContaBancaria:
    def __init__(self, titular, senha, saldo=0):
        self.__titular = titular
        self.__senha = senha
        self.__saldo = saldo

    def sacar(self, valor):
        if valor <= 0 or valor > self.__saldo:
            return False
        self.__saldo -= valor
        return True

    def depositar(self, valor):
        if valor <= 0:
            return False
        self.__saldo += valor
        return True

    def obter_saldo(self):
        return self.__saldo

    def esta_negativo(self):
        return self.__saldo < 0

    def trocar_senha(self, senha_atual, nova_senha):
        if senha_atual != self.__senha:
            return False
        self.__senha = nova_senha
        return True


joao = ContaBancaria('João', '1234')
maria = ContaBancaria('Maria', 'abcd', 250)

print(joao.depositar(100), joao.sacar(150), joao.sacar(40), joao.obter_saldo())
print(maria.sacar(-10), maria.depositar(0), maria.sacar(250), maria.obter_saldo())
print(maria.esta_negativo())
print(maria.trocar_senha('errada', '0000'), maria.trocar_senha('abcd', '0000'))
```

Os métodos que alteram o saldo retornam `True` ou `False` para quem usa a classe saber se
a operação deu certo. Na troca de senha, exigimos a senha atual: faz sentido no contexto
de um banco.

### Exercício 2

Crie uma classe para representar um **carro simplificado**. A classe deve guardar a
capacidade do tanque (em litros), a quantidade atual de gasolina e o consumo médio (em
km/litro). Implemente os métodos:

- a) `abastecer`: recebe uma quantidade de gasolina e a adiciona ao tanque. Atenção à
  capacidade do tanque e à quantidade atual. Retorna `True` ou `False`, conforme tenha
  sido possível abastecer.
- b) `informar_gasolina`: retorna a quantidade de gasolina disponível no tanque.
- c) `consumir_gasolina`: recebe uma distância em km e, usando o consumo médio, consome a
  quantidade correspondente de gasolina. Atenção: o tanque não pode ficar com quantidade
  negativa.
- d) `esta_na_reserva`: retorna um valor lógico indicando se a gasolina atual está abaixo
  de 10% da capacidade do tanque.

```python
class Carro:
    def __init__(self, capacidade, consumo_medio, gasolina=0):
        self.__capacidade = capacidade
        self.__consumo_medio = consumo_medio
        self.__gasolina = gasolina

    def abastecer(self, litros):
        if litros <= 0 or self.__gasolina + litros > self.__capacidade:
            return False
        self.__gasolina += litros
        return True

    def informar_gasolina(self):
        return self.__gasolina

    def consumir_gasolina(self, distancia):
        necessario = distancia / self.__consumo_medio
        if necessario > self.__gasolina:
            self.__gasolina = 0     # o carro anda até a gasolina acabar
            return False
        self.__gasolina -= necessario
        return True

    def esta_na_reserva(self):
        return self.__gasolina < 0.1 * self.__capacidade


carro = Carro(capacidade=50, consumo_medio=12)
print(carro.abastecer(60), carro.abastecer(40), carro.informar_gasolina())
print(carro.consumir_gasolina(420), carro.informar_gasolina(), carro.esta_na_reserva())
print(carro.consumir_gasolina(1000), carro.informar_gasolina())
```

Repare na chamada `Carro(capacidade=50, consumo_medio=12)`: informar o nome dos parâmetros
deixa claro o que cada número significa.

### Exercício 3

Crie uma classe `Pessoa` com três atributos (nome, rg e idade) e os métodos:

- a) construtor: com valores padrão para todos os atributos, ou seja, um objeto pode ser
  criado sem informar nenhum valor. Defina os valores padrão que achar mais adequados.
- b) `obter_nome`: retorna o nome da pessoa;
- c) `obter_idade`: retorna a idade da pessoa;
- d) `atualizar_idade`: recebe uma nova idade e altera o atributo (só se a nova idade for
  maior do que 0);
- e) `obter_rg`: retorna o RG da pessoa.

Crie também uma classe `CadastroPessoas`, que guarda objetos do tipo `Pessoa` em um
dicionário (analise as funcionalidades para decidir qual deve ser a chave). A classe deve:

- a) adicionar uma pessoa ao cadastro (a mesma pessoa não pode ser incluída duas vezes;
  use o RG para diferenciar as pessoas);
- b) consultar o nome e a idade de uma pessoa a partir do RG (se a pessoa não for
  encontrada, retorne vazio);
- c) remover uma pessoa do cadastro, informando o nome ou o RG, e retornar um valor lógico
  indicando se a remoção foi feita;
- d) informar qual é a pessoa com a maior idade;
- e) listar todas as pessoas cadastradas.

```python
class Pessoa:
    def __init__(self, nome='Sem nome', rg='', idade=0):
        self.__nome = nome
        self.__rg = rg
        self.__idade = idade

    def obter_nome(self):
        return self.__nome

    def obter_idade(self):
        return self.__idade

    def atualizar_idade(self, nova_idade):
        if nova_idade > 0:
            self.__idade = nova_idade

    def obter_rg(self):
        return self.__rg


class CadastroPessoas:
    def __init__(self):
        self.__pessoas = {}    # rg -> Pessoa (o RG identifica cada pessoa)

    def adicionar(self, pessoa):
        if pessoa.obter_rg() in self.__pessoas:
            return False
        self.__pessoas[pessoa.obter_rg()] = pessoa
        return True

    def consultar(self, rg):
        if rg not in self.__pessoas:
            return None
        pessoa = self.__pessoas[rg]
        return [pessoa.obter_nome(), pessoa.obter_idade()]

    def remover(self, nome_ou_rg):
        for rg, pessoa in self.__pessoas.items():
            if rg == nome_ou_rg or pessoa.obter_nome() == nome_ou_rg:
                del self.__pessoas[rg]
                return True    # sai do laço logo depois de remover
        return False

    def mais_velha(self):
        mais_velha = None
        for pessoa in self.__pessoas.values():
            if mais_velha is None or pessoa.obter_idade() > mais_velha.obter_idade():
                mais_velha = pessoa
        return mais_velha

    def listar(self):
        for pessoa in self.__pessoas.values():
            print(f'{pessoa.obter_nome()} (RG {pessoa.obter_rg()}), {pessoa.obter_idade()} anos')


cadastro = CadastroPessoas()
print(cadastro.adicionar(Pessoa('Ana', '111', 30)))
print(cadastro.adicionar(Pessoa('Bruno', '222', 45)))
print(cadastro.adicionar(Pessoa('Ana Clone', '111', 20)))   # RG repetido
print(cadastro.consultar('222'), cadastro.consultar('999'))
print(cadastro.mais_velha().obter_nome())
print(cadastro.remover('Ana'), cadastro.remover('Ninguém'))
cadastro.listar()
```

Em `remover`, retornamos logo depois do `del`. Isso é importante: alterar um dicionário
enquanto ele está sendo percorrido causa erro se o laço continuar.

### Exercício 4

Crie as classes necessárias para representar uma pequena **zona eleitoral** com 3 locais
de votação, cada um com uma urna eletrônica. Ao final da votação, a zona eleitoral deve
informar: o total de votos; o total e o percentual de votos de cada candidato (A, B e C);
o total e o percentual de votos nulos; o total e o percentual de votos brancos; e o
vencedor da eleição.

Cada urna permite que o eleitor vote em um dos candidatos, anule o voto ou vote em branco,
e informa os seus totais de votos.

```python
OPCOES = ['A', 'B', 'C', 'nulo', 'branco']


class Urna:
    def __init__(self):
        self.__votos = {}
        for opcao in OPCOES:
            self.__votos[opcao] = 0

    def votar(self, opcao):
        if opcao not in self.__votos:
            opcao = 'nulo'      # qualquer opção inválida anula o voto
        self.__votos[opcao] += 1

    def total(self, opcao):
        return self.__votos[opcao]


class ZonaEleitoral:
    def __init__(self, quantidade_locais=3):
        self.__urnas = []
        while len(self.__urnas) < quantidade_locais:
            self.__urnas.append(Urna())

    def urna(self, local):
        return self.__urnas[local]

    def total_opcao(self, opcao):
        soma = 0
        for urna in self.__urnas:
            soma += urna.total(opcao)
        return soma

    def total_votos(self):
        soma = 0
        for opcao in OPCOES:
            soma += self.total_opcao(opcao)
        return soma

    def vencedor(self):
        vencedor = 'A'
        for candidato in ['B', 'C']:
            if self.total_opcao(candidato) > self.total_opcao(vencedor):
                vencedor = candidato
        return vencedor

    def relatorio(self):
        total = self.total_votos()
        print(f'Total de votos: {total}')
        for opcao in OPCOES:
            votos = self.total_opcao(opcao)
            percentual = votos / total * 100 if total > 0 else 0
            print(f'{opcao}: {votos} votos ({percentual:.1f}%)')
        print(f'Vencedor: candidato {self.vencedor()}')


zona = ZonaEleitoral()
for voto in ['A', 'B', 'A', 'branco']:
    zona.urna(0).votar(voto)
for voto in ['C', 'A', 'xyz']:
    zona.urna(1).votar(voto)
for voto in ['B', 'A', 'nulo']:
    zona.urna(2).votar(voto)
zona.relatorio()
```

Cada classe tem uma responsabilidade: a `Urna` só registra e conta votos; a
`ZonaEleitoral` junta os resultados das urnas. Para simplificar, em caso de empate o
vencedor é o candidato que aparece primeiro.

A expressão `x if condicao else y` é um "if em uma linha": vale `x` quando a condição é
verdadeira e `y` caso contrário. Aqui ela evita a divisão por zero quando não há votos.
