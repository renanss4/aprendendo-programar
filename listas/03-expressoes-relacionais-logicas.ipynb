## Operadores relacionais

Um operador relacional **compara** dois valores e o resultado é sempre um valor lógico
(`bool`): `True` (verdadeiro) ou `False` (falso).

| Operador | Significado | Exemplo | Resultado |
| -------- | ----------- | ------- | --------- |
| `==` | igual a | `5 == 5` | `True` |
| `!=` | diferente de | `5 != 3` | `True` |
| `>` | maior que | `5 > 9` | `False` |
| `<` | menor que | `5 < 9` | `True` |
| `>=` | maior ou igual | `5 >= 5` | `True` |
| `<=` | menor ou igual | `6 <= 5` | `False` |

> **Atenção:** `=` atribui um valor a uma variável; `==` compara dois valores.

```python
x = 5
y = 10

print('x é igual a y?', x == y)
print('x é menor que y?', x < y)
print('x é diferente de y?', x != y)
```

## Operadores lógicos

Os operadores lógicos combinam valores `True`/`False`:

| Operador | Resultado é `True` quando... |
| -------- | ---------------------------- |
| `a and b` | **ambos** são verdadeiros |
| `a or b` | **pelo menos um** é verdadeiro |
| `not a` | `a` é falso (inverte o valor) |

```python
idade = 20
tem_carteira = True

pode_dirigir = idade >= 18 and tem_carteira
print('Pode dirigir?', pode_dirigir)

fim_de_semana = False
feriado = True
print('Dia de folga?', fim_de_semana or feriado)
print('Dia útil?', not (fim_de_semana or feriado))
```

## Intervalos

Para saber se `x` está entre 0 e 1, escreva `x > 0 and x < 1`. O Python também aceita a
forma encadeada `0 < x < 1`, igual à matemática.

## Dígitos com `//` e `%`

Combinando divisão inteira e resto, dá para "recortar" dígitos de um número:

```python
numero = 4729
print(numero % 10)          # último dígito: 9
print(numero // 10 % 10)    # penúltimo dígito: 2
print(numero // 100 % 10)   # antepenúltimo dígito: 7
```

## Erros comuns

- Usar `=` no lugar de `==` em comparações.
- Escrever `x > 0 and < 1`: cada lado do `and` precisa ser uma comparação completa
  (`x > 0 and x < 1`).
- Confundir `and` com `or`: "idade entre 22 e 40" é `idade >= 22 and idade <= 40`.
