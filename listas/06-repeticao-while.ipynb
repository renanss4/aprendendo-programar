## Teoria

## `while`: enquanto a condição for verdadeira

O `while` repete um bloco **enquanto** a condição for verdadeira. A cada volta (chamada de
**iteração**), a condição é testada de novo:

````python
contador = 1

while contador <= 5:
    print('Volta número', contador)
    contador = contador + 1

print('Fim!')
````

Todo `while` precisa de três partes:

1. **Inicialização:** `contador = 1`, antes do laço.
2. **Condição:** `contador <= 5`, testada no início de cada volta.
3. **Atualização:** `contador = contador + 1`, que faz a condição ficar falsa em algum
   momento.

> Se você esquecer a atualização, a condição nunca fica falsa e o programa roda para
> sempre (**laço infinito**). No notebook, interrompa clicando no botão de parar da célula.

`contador = contador + 1` também pode ser escrito como `contador += 1`.

## Contadores e acumuladores

- Um **contador** conta quantas vezes algo aconteceu (soma 1 a cada vez).
- Um **acumulador** junta valores ao longo do laço (por exemplo, uma soma ou um produto).

````python
# Soma dos números de 1 a 100 e quantidade de números pares
numero = 1
soma = 0      # acumulador
pares = 0     # contador

while numero <= 100:
    soma = soma + numero
    if numero % 2 == 0:
        pares = pares + 1
    numero = numero + 1

print('Soma:', soma)
print('Pares:', pares)
````

## Repetir até o usuário pedir para parar

Quando não sabemos quantas vezes vamos repetir, usamos um **valor sentinela**: um valor
especial que indica o fim. Repare que lemos o primeiro valor **antes** do laço e o próximo
no **final** de cada volta:

````python
valor = int(input('Digite um número (0 para sair): '))
soma = 0

while valor != 0:
    soma = soma + valor
    valor = int(input('Digite um número (0 para sair): '))

print('Soma:', soma)
````

## Teste de mesa

Quando o laço não fizer o que você espera, faça um **teste de mesa**: anote em uma tabela
o valor de cada variável a cada volta. Para o exemplo do contador:

| Volta | `contador` antes | Condição `contador <= 5` | O que mostra |
| ----- | ---------------- | ------------------------ | ------------ |
| 1 | 1 | `True` | Volta número 1 |
| 2 | 2 | `True` | Volta número 2 |
| ... | ... | ... | ... |
| 6 | 6 | `False` | (sai do laço) |

## Erros comuns

- Esquecer de atualizar a variável da condição (laço infinito).
- Inicializar o acumulador dentro do laço: ele volta a zero a cada iteração.
- Errar o limite: `< 10` para em 9, `<= 10` inclui o 10.
- Dividir pela quantidade sem verificar se ela é zero (por exemplo, ao calcular a média).

### Exercício 1

Mostre todos os números ímpares no intervalo de 0 até o número informado pelo usuário.
Mostre os ímpares apenas se a entrada for maior do que 1.

````python
limite = int(input('Digite um número: '))

if limite > 1:
    numero = 0
    while numero <= limite:
        if numero % 2 != 0:
            print(numero)
        numero += 1
else:
    print('O número precisa ser maior do que 1.')
````

### Exercício 4

Obtenha dois números inteiros que representam o limite inferior e o limite superior de um
somatório. Caso o limite inferior seja menor ou igual ao superior, calcule e mostre o
resultado do somatório. Caso contrário, avise o usuário e não faça o cálculo.

````python
inferior = int(input('Limite inferior: '))
superior = int(input('Limite superior: '))

if inferior <= superior:
    soma = 0
    numero = inferior
    while numero <= superior:
        soma += numero
        numero += 1
    print(f'Somatório de {inferior} até {superior}: {soma}')
else:
    print('O limite inferior precisa ser menor ou igual ao superior.')
````

### Exercício 5

Mostre os múltiplos de um número em ordem crescente. Obtenha, pela entrada do usuário, o
número cujos múltiplos serão mostrados e a quantidade de múltiplos. Comece a partir do 1
(ou seja, o primeiro múltiplo mostrado é o número vezes 1).

````python
numero = int(input('Número: '))
quantidade = int(input('Quantidade de múltiplos: '))

contador = 1
while contador <= quantidade:
    print(numero * contador)
    contador += 1
````

### Exercício 6

Obtenha um valor inteiro maior do que 0 e calcule o produtório de 1 até o valor
informado. Mostre o resultado do produtório e informe se ele é par ou ímpar.

````python
limite = int(input('Valor: '))

if limite > 0:
    produto = 1
    numero = 1
    while numero <= limite:
        produto *= numero
        numero += 1

    if produto % 2 == 0:
        print(f'Produtório: {produto} (par)')
    else:
        print(f'Produtório: {produto} (ímpar)')
else:
    print('O valor precisa ser maior do que 0.')
````

O produtório de 1 até `n` é o **fatorial** de `n` (escrito $n!$). Repare que o acumulador
de um produto começa em **1**, e não em 0: qualquer número multiplicado por 0 dá 0.

### Exercício 7

Leia números inteiros pela entrada do usuário até que ele digite o número 0. Informe a
quantidade de números digitados, o somatório e a média dos números (o 0 final não conta).

````python
numero = int(input('Número (0 para sair): '))
quantidade = 0
soma = 0

while numero != 0:
    quantidade += 1
    soma += numero
    numero = int(input('Número (0 para sair): '))

print(f'Quantidade: {quantidade}')
print(f'Somatório: {soma}')
if quantidade > 0:
    print(f'Média: {soma / quantidade:.2f}')
````

### Exercício 8

Obtenha um valor inteiro no intervalo de 1 a 9 (inclusive) e conte quantos números de 1 a
1000 (inclusive) são divisíveis pelo valor informado. Caso o usuário digite um valor fora
do intervalo de 1 a 9, avise-o e peça a entrada novamente. Repita até que o valor seja
válido ou que o usuário digite 0 (nesse caso, o programa termina sem fazer o cálculo).

````python
valor = int(input('Digite um valor de 1 a 9 (0 para sair): '))

while valor != 0 and (valor < 1 or valor > 9):
    print('Valor fora do intervalo de 1 a 9.')
    valor = int(input('Digite um valor de 1 a 9 (0 para sair): '))

if valor != 0:
    divisiveis = 0
    numero = 1
    while numero <= 1000:
        if numero % valor == 0:
            divisiveis += 1
        numero += 1
    print(f'{divisiveis} números de 1 a 1000 são divisíveis por {valor}.')
````

volta-se para A e o ciclo continua. Ao final, mostre quantos caracteres A, B e C foram
mostrados.

*Dica:* para mostrar os caracteres na mesma linha, vá concatenando (juntando) textos e
mostre o texto completo no final:

```python
resultado = 'A'
resultado = resultado + 'A'

print(resultado)  # AAB
```

Exemplo para a entrada 11: `A A A B B B C C C A A`

````python
quantidade = int(input('Quantidade de caracteres: '))

resultado = ''
letra = 'A'
seguidas = 0        # quantas vezes a letra atual já apareceu no ciclo
total_a = 0
total_b = 0
total_c = 0

contador = 0
while contador < quantidade:
    resultado = resultado + letra + ' '
    if letra == 'A':
        total_a += 1
    elif letra == 'B':
        total_b += 1
    else:
        total_c += 1

    seguidas += 1
    if seguidas == 3:
        seguidas = 0
        if letra == 'A':
            letra = 'B'
        elif letra == 'B':
            letra = 'C'
        else:
            letra = 'A'
    contador += 1

print(resultado)
print(f'A: {total_a}, B: {total_b}, C: {total_c}')
````
