## Criando listas novas

Para montar uma lista aos poucos, comece com uma lista vazia e use `append`, que adiciona
um elemento no final:

```python
numeros = [3, -1, 8, -5, 2]

positivos = []
for numero in numeros:
    if numero > 0:
        positivos.append(numero)

print(positivos)
```

## Erros comuns

- Começar a contar os índices em 1: o primeiro elemento é `lista[0]`.
- Usar `<= len(lista)` no `while`: o último índice válido é `len(lista) - 1`.
- Escrever `lista = lista.append(x)`: o `append` modifica a lista e devolve `None`.
  Escreva apenas `lista.append(x)`.
- Inicializar o "maior elemento" com 0: se todos os números forem negativos, o resultado
  fica errado. Comece com o primeiro elemento da lista.

### Exercício 1

Crie uma função que recebe a quantidade de números aleatórios desejada e retorna uma
lista com essa quantidade de números inteiros sorteados. Use alguma função do módulo
[`random`](https://docs.python.org/pt-br/3/library/random.html).

```python
from random import randint

def numeros_aleatorios(quantidade):
    numeros = []
    while len(numeros) < quantidade:
        numeros.append(randint(1, 100))
    return numeros

print(numeros_aleatorios(5))
```

### Exercício 2

Crie uma função que recebe uma lista de números e retorna uma **nova** lista com o módulo
(valor absoluto) de cada elemento da lista original.

```python
def modulos(lista):
    resultado = []
    for numero in lista:
        if numero < 0:
            resultado.append(-numero)
        else:
            resultado.append(numero)
    return resultado

print(modulos([-5, 10, -15, 0]))
```

### Exercício 3

Crie uma função que recebe uma lista de números, encontra o maior deles e retorna uma
nova lista com todos os números da lista original divididos pelo maior. Use a função
[`max`](https://docs.python.org/pt-br/3/library/functions.html#max) do Python.

```python
def dividir_pelo_maior(lista):
    maior = max(lista)
    resultado = []
    for numero in lista:
        resultado.append(numero / maior)
    return resultado

print(dividir_pelo_maior([2, 4, 8, 10]))
```

### Exercício 4

Crie uma função que recebe uma lista de números e retorna uma nova lista com os números
primos dessa lista. Use uma função para a verificação de primos, assim como no exercício 2 da lista 9.

```python
def eh_primo(numero):
    if numero < 2:
        return False
    divisor = 2
    while divisor * divisor <= numero:
        if numero % divisor == 0:
            return False
        divisor += 1
    return True


def filtrar_primos(lista):
    primos = []
    for numero in lista:
        if eh_primo(numero):
            primos.append(numero)
    return primos

print(filtrar_primos([0, 1, 2, 3, 5, 6, 29, 30]))
```

### Exercício 5

Crie uma função que recebe uma quantidade `N` e um número de início, e retorna uma lista
com os `N` primeiros números primos a partir do número recebido. Crie e use uma
função `proximo_primo`, que retorna o menor número primo maior do que o número recebido.

```python
def proximo_primo(numero):
    candidato = numero + 1
    while not eh_primo(candidato):
        candidato += 1
    return candidato

def primos_a_partir_de(n, inicio):
    primos = []
    atual = inicio - 1   # assim, o próprio início entra se for primo
    while len(primos) < n:
        atual = proximo_primo(atual)
        primos.append(atual)
    return primos

print(primos_a_partir_de(5, 10))
print(primos_a_partir_de(3, 7))
```

### Exercício 6

Crie uma função que recebe duas listas e retorna uma nova lista com a multiplicação dos
elementos de mesmo índice das listas de entrada. Caso as listas tenham tamanhos
diferentes, inclua diretamente os elementos restantes da lista maior.

Exemplo: `[1, 2, 3, 4]` e `[10, 10]` → `[10, 20, 3, 4]`

```python
def multiplicar_listas(lista1, lista2):
    resultado = []
    indice = 0
    while indice < len(lista1) or indice < len(lista2):
        if indice < len(lista1) and indice < len(lista2):
            resultado.append(lista1[indice] * lista2[indice])
        elif indice < len(lista1):
            resultado.append(lista1[indice])
        else:
            resultado.append(lista2[indice])
        indice += 1
    return resultado

print(multiplicar_listas([1, 2, 3, 4], [10, 10]))
```

### Exercício 7

Crie uma função que recebe uma lista de tamanho par e maior do que 0 (caso não seja,
retorne vazio) e troca o primeiro elemento com o último, o segundo com o penúltimo, o
terceiro com o antepenúltimo, e assim por diante. Retorne uma **nova** lista com as trocas.

```python
def trocar_extremos(lista):
    if len(lista) == 0 or len(lista) % 2 != 0:
        return None

    resultado = []
    for elemento in lista:
        resultado.append(elemento)   # copia a lista para não alterar a original

    inicio = 0
    fim = len(resultado) - 1
    while inicio < fim:
        temporario = resultado[inicio]
        resultado[inicio] = resultado[fim]
        resultado[fim] = temporario
        inicio += 1
        fim -= 1
    return resultado

original = [1, 2, 3, 4, 5, 6]
print(trocar_extremos(original))
print(original)
print(trocar_extremos([1, 2, 3]))
```

Para trocar dois valores, guardamos um deles em uma variável temporária. Sem ela, o
primeiro valor seria perdido ao ser sobrescrito.

### Exercício 8

Crie uma função que recebe duas listas e retorna uma nova lista apenas com os elementos
presentes em **ambas** as listas.

```python
def elementos_em_comum(lista1, lista2):
    comuns = []
    for elemento in lista1:
        if elemento in lista2 and elemento not in comuns:
            comuns.append(elemento)
    return comuns

print(elementos_em_comum([1, 2, 3, 4, 5, 3], [3, 4, 5, 6, 7]))
```

O teste `elemento not in comuns` evita repetir um elemento que aparece mais de uma vez na
primeira lista.
