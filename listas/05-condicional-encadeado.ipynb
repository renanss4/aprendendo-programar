## Teoria

## `elif`: várias condições

Quando há mais de dois caminhos, use `elif` (senão, se). O Python testa as condições
**em ordem** e executa apenas o bloco da **primeira** que for verdadeira:

```python
nota = 7.5

if nota >= 9:
    print('Conceito A')
elif nota >= 7:
    print('Conceito B')
elif nota >= 5:
    print('Conceito C')
else:
    print('Reprovado')
```

Como a nota 7.5 não passou no primeiro teste (`>= 9`), o Python testou o segundo
(`>= 7`), que é verdadeiro, mostrou "Conceito B" e **pulou todo o resto**.

## Condicionais aninhadas

Um `if` pode ficar dentro de outro. Isso é útil para validar a entrada antes de calcular:

```python
divisor = 0
dividendo = 10

if divisor != 0:
    resultado = dividendo / divisor
    if resultado > 1:
        print('O resultado é maior que 1')
    else:
        print('O resultado é no máximo 1')
else:
    print('Não é possível dividir por zero!')
```

## Erros comuns

- Esquecer os dois-pontos depois do `if`, `elif` ou `else`.
- Indentação inconsistente: misturar 2 e 4 espaços causa `IndentationError`.
- Usar vários `if` independentes quando os casos são exclusivos. Com `if`/`elif`, só um
  bloco é executado; com vários `if`, todos são testados.
- Esquecer que a ordem do `elif` importa: no exemplo das notas, se o teste `>= 5` viesse
  primeiro, a nota 9.5 receberia "Conceito C".

### Exercício 1

Obtenha três valores numéricos pela entrada do usuário, representando os ângulos de um
triângulo. Caso a soma dos ângulos internos ultrapasse 180°, avise o usuário e encerre o
programa. A partir dos ângulos, classifique o triângulo e mostre a categoria:

- **Acutângulo:** todos os ângulos com medidas menores que 90°;
- **Retângulo:** um ângulo com medida igual a 90°;
- **Obtusângulo:** um ângulo obtuso, maior que 90°.

```python
angulo1 = float(input('Ângulo 1: '))
angulo2 = float(input('Ângulo 2: '))
angulo3 = float(input('Ângulo 3: '))

if angulo1 + angulo2 + angulo3 > 180:
    print('A soma dos ângulos ultrapassa 180°.')
elif angulo1 == 90 or angulo2 == 90 or angulo3 == 90:
    print('Triângulo retângulo')
elif angulo1 > 90 or angulo2 > 90 or angulo3 > 90:
    print('Triângulo obtusângulo')
else:
    print('Triângulo acutângulo')
```

### Exercício 3

Uma loja virtual dá desconto aos clientes com base no seu nível de engajamento nas redes
sociais da loja. Obtenha o nível de engajamento do cliente e o valor da compra (o valor
precisa ser superior a R\$ 100,00). Os níveis aceitos e os percentuais de desconto são:

| Nível | Desconto |
| ----- | -------- |
| seguidor | 5% |
| comentarista | 8% |
| fã | 12% |

Caso o nível informado não seja nenhum desses ou o valor não satisfaça a condição, mostre
uma mensagem avisando que o cliente não tem direito a desconto. Caso contrário, mostre o
percentual de desconto e o valor final da compra.

```python
nivel = input('Nível de engajamento (seguidor, comentarista ou fã): ')
valor = float(input('Valor da compra: R$ '))

if valor <= 100:
    percentual = 0
elif nivel == 'seguidor':
    percentual = 5
elif nivel == 'comentarista':
    percentual = 8
elif nivel == 'fã':
    percentual = 12
else:
    percentual = 0

if percentual == 0:
    print('O cliente não tem direito a desconto.')
else:
    valor_final = valor - valor * percentual / 100
    print(f'Desconto de {percentual}%. Valor final: R${valor_final:.2f}')
```

### Exercício 4

Converta uma temperatura informada pelo usuário para as duas outras unidades de medida.
O usuário informa uma unidade (Celsius, Fahrenheit ou Kelvin) e um valor de temperatura;
o programa mostra a conversão para as outras duas. Caso o usuário informe uma unidade
diferente dessas, avise-o. Conversões:

- $°C = (°F - 32) / 1{,}8$
- $°C = K - 273$
- $°F = (K - 273) \times 1{,}8 + 32$

```python
unidade = input('Unidade (Celsius, Fahrenheit ou Kelvin): ')
temperatura = float(input('Temperatura: '))

if unidade == 'Celsius':
    fahrenheit = temperatura * 1.8 + 32
    kelvin = temperatura + 273
    print(f'{fahrenheit:.1f}°F e {kelvin:.1f}K')
elif unidade == 'Fahrenheit':
    celsius = (temperatura - 32) / 1.8
    kelvin = celsius + 273
    print(f'{celsius:.1f}°C e {kelvin:.1f}K')
elif unidade == 'Kelvin':
    celsius = temperatura - 273
    fahrenheit = (temperatura - 273) * 1.8 + 32
    print(f'{celsius:.1f}°C e {fahrenheit:.1f}°F')
else:
    print('Unidade inválida.')
```

As fórmulas do enunciado convertem **para** Celsius. Para converter **a partir** de
Celsius, basta "inverter" cada fórmula: $°F = °C \times 1{,}8 + 32$ e $K = °C + 273$.

### Exercício 5

Obtenha três valores numéricos pela entrada do usuário, representando os lados de um
triângulo. Informe se o triângulo é **equilátero** (3 lados iguais), **isósceles**
(2 lados iguais) ou **escaleno** (nenhum lado igual). Caso algum lado seja negativo, ou
as medidas não obedeçam à condição de existência de um triângulo, avise o usuário e não
faça a classificação.

> **Condição de existência:** cada lado deve ser maior que o valor absoluto da diferença
> dos outros dois lados e menor que a soma dos outros dois.

```python
a = float(input('Lado 1: '))
b = float(input('Lado 2: '))
c = float(input('Lado 3: '))

existe = (abs(b - c) < a < b + c) and (abs(a - c) < b < a + c) and (abs(a - b) < c < a + b)

if a < 0 or b < 0 or c < 0:
    print('Os lados não podem ser negativos.')
elif not existe:
    print('Essas medidas não formam um triângulo.')
elif a == b and b == c:
    print('Triângulo equilátero')
elif a == b or a == c or b == c:
    print('Triângulo isósceles')
else:
    print('Triângulo escaleno')
```

### Exercício 6

O estado físico da água depende da pressão e da temperatura. Obtenha um valor de pressão
$p$ (em kPa) e uma temperatura $T$ (em °C) e informe em qual estado a água se encontra,
considerando as aproximações abaixo:

- **p = 100 kPa:** sólido se T < 0 °C; mistura sólido-líquido se T = 0 °C; líquido se
  0 °C < T < 100 °C; mistura líquido-vapor se T = 100 °C; gasoso se T > 100 °C.
- **p = 200 kPa:** mistura líquido-vapor se T = 120 °C; gasoso se T > 120 °C.
- **p = 300 kPa:** mistura líquido-vapor se T = 133,6 °C; gasoso se T > 133,6 °C.
- **p = 400 kPa:** mistura líquido-vapor se T = 143,6 °C; gasoso se T > 143,6 °C.
- **p = 500 kPa:** mistura líquido-vapor se T = 151,9 °C; gasoso se T > 151,9 °C.

Se o usuário informar valores que não constam nessa tabela, mostre uma mensagem dizendo
que os valores não constam no banco de dados.

```python
pressao = float(input('Pressão (kPa): '))
temperatura = float(input('Temperatura (°C): '))

estado = 'Os valores informados não constam no banco de dados.'

if pressao == 100:
    if temperatura < 0:
        estado = 'Sólido'
    elif temperatura == 0:
        estado = 'Mistura sólido-líquido'
    elif temperatura < 100:
        estado = 'Líquido'
    elif temperatura == 100:
        estado = 'Mistura líquido-vapor'
    else:
        estado = 'Gasoso'
else:
    # Para as outras pressões, só muda a temperatura de ebulição
    ebulicao = 0
    if pressao == 200:
        ebulicao = 120
    elif pressao == 300:
        ebulicao = 133.6
    elif pressao == 400:
        ebulicao = 143.6
    elif pressao == 500:
        ebulicao = 151.9

    if ebulicao != 0:
        if temperatura == ebulicao:
            estado = 'Mistura líquido-vapor'
        elif temperatura > ebulicao:
            estado = 'Gasoso'

print(estado)
```

Repare que, para as pressões de 200 a 500 kPa, só a temperatura de ebulição muda. Em vez
de repetir o mesmo bloco quatro vezes, guardamos essa temperatura em uma variável e
fazemos a comparação uma vez só.
