## Matriz: uma lista de listas

Uma **matriz** é uma tabela de valores organizada em linhas e colunas. Em Python, ela é
representada por uma lista em que cada elemento é uma linha (que também é uma lista):

````python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
]

print(len(matriz))      # quantidade de linhas: 2
print(len(matriz[0]))   # quantidade de colunas: 3
print(matriz[1])        # a linha 1 inteira: [4, 5, 6]
print(matriz[1][2])     # linha 1, coluna 2: 6
````

O acesso é sempre `matriz[linha][coluna]`, e os índices começam em 0:

|  | coluna 0 | coluna 1 | coluna 2 |
| - | - | - | - |
| **linha 0** | `m[0][0]` = 1 | `m[0][1]` = 2 | `m[0][2]` = 3 |
| **linha 1** | `m[1][0]` = 4 | `m[1][1]` = 5 | `m[1][2]` = 6 |

## Percorrendo uma matriz

Usamos dois laços aninhados: o de fora percorre as linhas e o de dentro, as colunas:

````python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
]

linha = 0
while linha < len(matriz):
    coluna = 0
    while coluna < len(matriz[linha]):
        print(matriz[linha][coluna], end=' ')
        coluna += 1
    print()
    linha += 1
````

Para percorrer **coluna a coluna** (por exemplo, para somar cada coluna), basta inverter
os laços: o de fora percorre as colunas e o de dentro, as linhas.

Quando não precisar dos índices, o `for` também funciona:

````python
soma = 0
for linha in matriz:
    for elemento in linha:
        soma += elemento
print('Soma:', soma)
````

## Erros comuns

- Inverter linha e coluna: é `m[linha][coluna]`.
- Usar `len(matriz)` como quantidade de colunas: isso é a quantidade de **linhas**.
  Colunas são `len(matriz[0])`.
- Criar a matriz com `[[0] * colunas] * linhas` (as linhas ficam ligadas).
- Esquecer de reiniciar o índice da coluna a cada nova linha.

### Exercício 1

Crie uma função que recebe uma matriz numérica e retorna o somatório de todos os
elementos.

````python
def soma_matriz(matriz):
    soma = 0
    for linha in matriz:
        for elemento in linha:
            soma += elemento
    return soma

print(soma_matriz([[1, 2, 3], [4, 5, 6]]))
````

### Exercício 2

Crie uma função que recebe uma matriz e a mostra na forma de linhas x colunas, com os
elementos separados por um espaço em branco.

````python
def mostrar_matriz(matriz):
    for linha in matriz:
        for elemento in linha:
            print(elemento, end=' ')
        print()

mostrar_matriz([[1, 2, 3], [4, 5, 6]])
````

### Exercício 3

Crie uma função que recebe uma matriz de textos e retorna o maior texto (o de maior
tamanho) armazenado na matriz.

````python
def maior_texto(matriz):
    maior = matriz[0][0]
    for linha in matriz:
        for texto in linha:
            if len(texto) > len(maior):
                maior = texto
    return maior

print(maior_texto([['sol', 'lua'], ['estrela', 'mar']]))
````

### Exercício 4

Crie uma função que recebe uma matriz numérica e retorna uma lista com o menor elemento
da matriz e a sua posição (linha e coluna).

````python
def menor_elemento(matriz):
    menor = matriz[0][0]
    linha_menor = 0
    coluna_menor = 0

    linha = 0
    while linha < len(matriz):
        coluna = 0
        while coluna < len(matriz[linha]):
            if matriz[linha][coluna] < menor:
                menor = matriz[linha][coluna]
                linha_menor = linha
                coluna_menor = coluna
            coluna += 1
        linha += 1

    return [menor, linha_menor, coluna_menor]

print(menor_elemento([[100, 22, 32], [212, 15, 1], [35, 24, 21]]))
````

### Exercício 5

Crie uma função que recebe uma matriz numérica e retorna quantos elementos são números
primos. Use uma função `eh_primo` para a verificação.

````python
def eh_primo(numero):
    if numero < 2:
        return False
    divisor = 2
    while divisor * divisor <= numero:
        if numero % divisor == 0:
            return False
        divisor += 1
    return True

def contar_primos_matriz(matriz):
    quantidade = 0
    for linha in matriz:
        for elemento in linha:
            if eh_primo(elemento):
                quantidade += 1
    return quantidade

print(contar_primos_matriz([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))
````

### Exercício 6

Crie uma função que recebe uma matriz numérica e um número, e retorna um valor lógico
indicando se a matriz contém o número.

````python
def matriz_contem(matriz, numero):
    for linha in matriz:
        for elemento in linha:
            if elemento == numero:
                return True
    return False

print(matriz_contem([[1, 2], [3, 4]], 3))
print(matriz_contem([[1, 2], [3, 4]], 9))
````

### Exercício 7

Crie uma função que recebe uma matriz numérica e retorna um valor lógico indicando se ela
é **quadrada** (mesma quantidade de linhas e colunas, como 5x5 ou 10x10).

````python
def eh_quadrada(matriz):
    for linha in matriz:
        if len(linha) != len(matriz):
            return False
    return True

print(eh_quadrada([[1, 2], [3, 4]]))
print(eh_quadrada([[1, 2, 3], [4, 5, 6]]))
````

Verificamos **todas** as linhas: uma matriz como `[[1, 2], [3]]` tem a primeira linha com
o tamanho certo, mas não é quadrada.

### Exercício 8

Crie uma função que recebe uma matriz numérica quadrada (verifique usando a função do
exercício 7) e retorna uma lista com o somatório dos elementos de cada coluna. Se a matriz
não for quadrada, retorne vazio.

````python
def soma_colunas(matriz):
    if not eh_quadrada(matriz):
        return None

    somas = []
    coluna = 0
    while coluna < len(matriz[0]):
        soma = 0
        linha = 0
        while linha < len(matriz):
            soma += matriz[linha][coluna]
            linha += 1
        somas.append(soma)
        coluna += 1
    return somas

print(soma_colunas([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))
````

Para somar por coluna, o laço de fora percorre as **colunas** e o de dentro, as linhas.

### Exercício 9

Crie uma função que recebe uma matriz de textos e informa quais linhas e quais colunas
possuem valores vazios (`''`).

````python
def mostrar_vazios(matriz):
    linhas_com_vazio = []
    colunas_com_vazio = []

    linha = 0
    while linha < len(matriz):
        coluna = 0
        while coluna < len(matriz[linha]):
            if matriz[linha][coluna] == '':
                if linha not in linhas_com_vazio:
                    linhas_com_vazio.append(linha)
                if coluna not in colunas_com_vazio:
                    colunas_com_vazio.append(coluna)
            coluna += 1
        linha += 1

    print('Linhas com valores vazios:', linhas_com_vazio)
    print('Colunas com valores vazios:', colunas_com_vazio)

mostrar_vazios([['a', 'b', ''], ['', 'c', 'd'], ['e', 'f', 'g']])
````

### Exercício 10

Crie uma função que recebe uma matriz numérica e informa quais linhas e quais colunas
possuem **somente** valores negativos.

````python
def mostrar_negativos(matriz):
    linha = 0
    while linha < len(matriz):
        todos_negativos = True
        for elemento in matriz[linha]:
            if elemento >= 0:
                todos_negativos = False
        if todos_negativos:
            print(f'A linha {linha} tem somente valores negativos')
        linha += 1

    coluna = 0
    while coluna < len(matriz[0]):
        todos_negativos = True
        linha = 0
        while linha < len(matriz):
            if matriz[linha][coluna] >= 0:
                todos_negativos = False
            linha += 1
        if todos_negativos:
            print(f'A coluna {coluna} tem somente valores negativos')
        coluna += 1

mostrar_negativos([[1, -2, 4], [3, -5, 6], [-8, -7, -9]])
````

A variável `todos_negativos` começa como `True` e vira `False` assim que um elemento não
negativo aparece. Esse padrão ("supor verdadeiro até encontrar um contraexemplo") é muito
útil.

### Exercício 11

Crie uma função que recebe uma matriz quadrada e retorna um valor lógico indicando se ela
é **simétrica**. Uma matriz é simétrica quando, para toda linha `l` e coluna `c`,
`M[l][c] == M[c][l]`.

````python
def eh_simetrica(matriz):
    linha = 0
    while linha < len(matriz):
        coluna = 0
        while coluna < len(matriz):
            if matriz[linha][coluna] != matriz[coluna][linha]:
                return False
            coluna += 1
        linha += 1
    return True

print(eh_simetrica([[1, 2, 4], [2, 5, 6], [4, 6, 9]]))
print(eh_simetrica([[1, 2], [3, 4]]))
````

### Exercício 12

O elemento **MINMAX** de uma matriz é o maior elemento da linha em que se encontra o menor
elemento da matriz. Crie uma função que recebe uma matriz e retorna uma lista com
3 elementos: a linha, a coluna e o valor do elemento minmax.

````python
def minmax(matriz):
    # 1) Encontra a linha do menor elemento (reaproveita a função do exercício 4)
    linha = menor_elemento(matriz)[1]

    # 2) Encontra o maior elemento dessa linha
    coluna_maior = 0
    coluna = 0
    while coluna < len(matriz[linha]):
        if matriz[linha][coluna] > matriz[linha][coluna_maior]:
            coluna_maior = coluna
        coluna += 1

    return [linha, coluna_maior, matriz[linha][coluna_maior]]

print(minmax([[5, 8, 3], [9, 1, 7], [4, 6, 2]]))
````

### Exercício 13

Um elemento `M[lin][col]` é um **ponto de sela** da matriz `M` se, e somente se, ele for
ao mesmo tempo o **menor** elemento da linha `lin` e o **maior** elemento da coluna `col`.
Crie uma função que recebe uma matriz e informa se ela possui um ponto de sela e, se
possuir, mostra o seu valor e a sua localização.

````python
def ponto_de_sela(matriz):
    encontrou = False
    linha = 0
    while linha < len(matriz):
        coluna = 0
        while coluna < len(matriz[linha]):
            valor = matriz[linha][coluna]

            menor_da_linha = True
            for elemento in matriz[linha]:
                if elemento < valor:
                    menor_da_linha = False

            maior_da_coluna = True
            for outra_linha in matriz:
                if outra_linha[coluna] > valor:
                    maior_da_coluna = False

            if menor_da_linha and maior_da_coluna:
                print(f'Ponto de sela: {valor} na linha {linha}, coluna {coluna}')
                encontrou = True
            coluna += 1
        linha += 1

    if not encontrou:
        print('A matriz não possui ponto de sela.')

ponto_de_sela([[3, 8, 5], [1, 0, 6], [2, 4, 9]])
ponto_de_sela([[1, 2], [2, 1]])
````

### Exercício 14

Crie uma função que recebe uma matriz numérica e retorna uma lista com todos os elementos
da matriz, sem repetições. Exemplo:

```
[[1, 2, 2],
 [3, 1, 1]]  →  [1, 2, 3]
```

````python
def elementos_unicos(matriz):
    unicos = []
    for linha in matriz:
        for elemento in linha:
            if elemento not in unicos:
                unicos.append(elemento)
    return unicos

print(elementos_unicos([[1, 2, 2], [3, 1, 1]]))
````

### Exercício 15

Crie uma função que recebe uma matriz numérica, verifica quais elementos estão repetidos
e quantas vezes cada um aparece. Mostre ao usuário quantas vezes cada elemento aparece na
matriz.

````python
def mostrar_repeticoes(matriz):
    elementos = []      # elementos distintos, na ordem em que aparecem
    quantidades = []    # quantidades[i] é quantas vezes elementos[i] aparece

    for linha in matriz:
        for valor in linha:
            if valor in elementos:
                indice = elementos.index(valor)
                quantidades[indice] += 1
            else:
                elementos.append(valor)
                quantidades.append(1)

    indice = 0
    while indice < len(elementos):
        print(f'{elementos[indice]} aparece {quantidades[indice]} vez(es)')
        indice += 1

mostrar_repeticoes([[1, 2, 2], [3, 1, 1]])
````

Usamos duas listas "paralelas": a posição `i` de uma corresponde à posição `i` da outra.
O método `lista.index(valor)` devolve a posição de `valor` na lista. Na lista 14 você
vai ver que um **dicionário** resolve esse problema de forma mais natural.
