## Criando uma matriz

Crie cada linha como uma lista nova e adicione-a à matriz com `append`:

```python
linhas = 3
colunas = 4

matriz = []
i = 0
while i < linhas:
    linha = []
    j = 0
    while j < colunas:
        linha.append(i * 10 + j)
        j += 1
    matriz.append(linha)
    i += 1

print(matriz)
```

> **Cuidado:** `[[0] * 3] * 2` parece criar uma matriz 2x3, mas as duas linhas são a
> **mesma** lista! Alterar `m[0][0]` altera também `m[1][0]`. Crie cada linha
> separadamente, como no exemplo acima.

## Erros comuns

- Inverter linha e coluna: é `m[linha][coluna]`.
- Usar `len(matriz)` como quantidade de colunas: isso é a quantidade de **linhas**.
  Colunas são `len(matriz[0])`.
- Criar a matriz com `[[0] * colunas] * linhas` (as linhas ficam ligadas).
- Esquecer de reiniciar o índice da coluna a cada nova linha.

### Exercício 1

Crie uma função que recebe uma quantidade de linhas e uma de colunas, e retorna uma
matriz com essas dimensões em que todos os elementos valem 0.

```python
def matriz_de_zeros(linhas, colunas):
    matriz = []
    while len(matriz) < linhas:
        linha = []
        while len(linha) < colunas:
            linha.append(0)
        matriz.append(linha)
    return matriz

print(matriz_de_zeros(2, 3))
```

### Exercício 2

Crie uma função que recebe uma quantidade de linhas, uma de colunas e um limite (inteiro),
e retorna uma matriz com essas dimensões em que todos os elementos são números aleatórios
no intervalo de `-limite` a `+limite`.

```python
from random import randint

def matriz_aleatoria(linhas, colunas, limite):
    matriz = []
    while len(matriz) < linhas:
        linha = []
        while len(linha) < colunas:
            linha.append(randint(-limite, limite))
        matriz.append(linha)
    return matriz

print(matriz_aleatoria(3, 3, 5))
```

### Exercício 3

Crie uma função que recebe uma matriz e retorna uma lista com a quantidade de linhas e de
colunas da matriz. Caso alguma linha tenha uma quantidade de colunas diferente da
primeira, retorne vazio.

```python
def dimensoes(matriz):
    colunas = len(matriz[0])
    for linha in matriz:
        if len(linha) != colunas:
            return None
    return [len(matriz), colunas]

print(dimensoes([[1, 2, 3], [4, 5, 6]]))
print(dimensoes([[1, 2, 3], [4, 5]]))
```

### Exercício 4

Crie uma função que recebe duas matrizes numéricas e retorna uma nova matriz com a soma
dos elementos de mesmo índice. Caso as matrizes não tenham as mesmas dimensões, retorne
`False`. Use a função do exercício 3 para essa verificação.

```python
def somar_matrizes(m1, m2):
    if dimensoes(m1) is None or dimensoes(m1) != dimensoes(m2):
        return False

    resultado = []
    linha = 0
    while linha < len(m1):
        nova_linha = []
        coluna = 0
        while coluna < len(m1[linha]):
            nova_linha.append(m1[linha][coluna] + m2[linha][coluna])
            coluna += 1
        resultado.append(nova_linha)
        linha += 1
    return resultado

print(somar_matrizes([[1, 2], [3, 4]], [[10, 20], [30, 40]]))
print(somar_matrizes([[1, 2], [3, 4]], [[1, 2]]))
```

### Exercício 5

Crie uma função que recebe uma matriz numérica e retorna uma nova matriz em que todos os
elementos são multiplicados por 10, com exceção da diagonal principal, cujos elementos
devem ser 0.

```python
def multiplicar_fora_da_diagonal(matriz):
    resultado = []
    linha = 0
    while linha < len(matriz):
        nova_linha = []
        coluna = 0
        while coluna < len(matriz[linha]):
            if linha == coluna:
                nova_linha.append(0)
            else:
                nova_linha.append(matriz[linha][coluna] * 10)
            coluna += 1
        resultado.append(nova_linha)
        linha += 1
    return resultado

print(multiplicar_fora_da_diagonal([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))
```

### Exercício 6

Crie uma função que recebe uma matriz numérica com três colunas (não é necessário
verificar) e retorna uma nova matriz com as seguintes modificações:

- a) 1ª coluna: multiplicar todos os elementos por 10;
- b) 2ª coluna: calcular o fatorial de cada elemento;
- c) 3ª coluna: o módulo (valor absoluto) de cada elemento.

Use a função `factorial` do módulo `math` e a função `abs` do Python.

```python
from math import factorial

def transformar_colunas(matriz):
    resultado = []
    for linha in matriz:
        resultado.append([linha[0] * 10, factorial(linha[1]), abs(linha[2])])
    return resultado

print(transformar_colunas([[1, 5, -3], [5, 4, 3], [10, 1, -4]]))
```

### Exercício 7

Crie uma função que recebe uma matriz quadrada e retorna uma matriz com duas linhas: na
primeira, os **maiores** elementos de cada linha da matriz de entrada; na segunda, os
**menores** elementos de cada coluna. Para verificar se a matriz é quadrada, reaproveite
a função do exercício 7 da lista 11. *(Adaptado: no original, a função era importada de um arquivo
`.py`.)*

```python
def eh_quadrada(matriz):
    for linha in matriz:
        if len(linha) != len(matriz):
            return False
    return True


def maiores_e_menores(matriz):
    if not eh_quadrada(matriz):
        return None

    n = len(matriz)
    maiores_linhas = []
    menores_colunas = []

    i = 0
    while i < n:
        maior = matriz[i][0]
        menor = matriz[0][i]
        j = 0
        while j < n:
            if matriz[i][j] > maior:    # percorre a linha i
                maior = matriz[i][j]
            if matriz[j][i] < menor:    # percorre a coluna i
                menor = matriz[j][i]
            j += 1
        maiores_linhas.append(maior)
        menores_colunas.append(menor)
        i += 1

    return [maiores_linhas, menores_colunas]

print(maiores_e_menores([[3, 6, 2], [8, 1, 7], [4, 5, 9]]))
```

Como a matriz é quadrada, o mesmo par de laços serve para percorrer a linha `i`
(`matriz[i][j]`) e a coluna `i` (`matriz[j][i]`) ao mesmo tempo.

### Exercício 8

Crie uma função que recebe uma matriz quadrada e retorna uma nova matriz em que cada
elemento de uma linha é multiplicado pelo elemento da **diagonal secundária** daquela
linha.

```python
def multiplicar_pela_secundaria(matriz):
    n = len(matriz)
    resultado = []
    linha = 0
    while linha < n:
        fator = matriz[linha][n - 1 - linha]
        nova_linha = []
        for elemento in matriz[linha]:
            nova_linha.append(elemento * fator)
        resultado.append(nova_linha)
        linha += 1
    return resultado

print(multiplicar_pela_secundaria([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))
```
