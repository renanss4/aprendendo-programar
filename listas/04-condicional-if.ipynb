## Teoria

## `if`: tomando decisões

Até agora, todos os programas executavam todas as linhas, de cima para baixo. Com o `if`,
um bloco de código só é executado **se** a condição for verdadeira:

```python
idade = 20

if idade >= 18:
    print('Você é maior de idade.')
print('Esta linha sempre aparece.')
```

Repare em dois detalhes:

1. A linha do `if` termina com **dois-pontos** (`:`).
2. O bloco que pertence ao `if` fica **indentado** (recuado 4 espaços). É a indentação que
   diz ao Python onde o bloco começa e termina.

## `if` / `else`: dois caminhos

O `else` (senão) define o que fazer quando a condição é falsa:

```python
numero = 7

if numero % 2 == 0:
    print(f'{numero} é par')
else:
    print(f'{numero} é ímpar')
```

## Erros comuns

- Esquecer os dois-pontos depois do `if`, `elif` ou `else`.
- Indentação inconsistente: misturar 2 e 4 espaços causa `IndentationError`.
- Usar vários `if` independentes quando os casos são exclusivos. Com `if`/`elif`, só um
  bloco é executado; com vários `if`, todos são testados.
- Esquecer que a ordem do `elif` importa: no exemplo das notas, se o teste `>= 5` viesse
  primeiro, a nota 9.5 receberia "Conceito C".

### Exercício 1

Calcule e mostre a hipotenusa de um triângulo retângulo, obtendo os valores dos catetos
pela entrada do usuário com a mensagem `Informe o valor do cateto N:` (onde `N` indica
qual cateto deve ser informado). **Caso algum valor informado seja negativo, mostre uma
mensagem de aviso e não faça o cálculo.** Para mostrar o resultado, use a mensagem
`A hipotenusa é: R`.

$$hip = \sqrt{cat_1^2 + cat_2^2}$$

```python
cateto1 = float(input('Informe o valor do cateto 1: '))
cateto2 = float(input('Informe o valor do cateto 2: '))

if cateto1 < 0 or cateto2 < 0:
    print('Os catetos não podem ser negativos.')
else:
    hipotenusa = (cateto1 ** 2 + cateto2 ** 2) ** (1 / 2)
    print(f'A hipotenusa é: {hipotenusa}')
```

### Exercício 5

Peça ao usuário dois valores numéricos e informe se o primeiro valor é múltiplo do
segundo.

```python
num1 = int(input('Primeiro valor: '))
num2 = int(input('Segundo valor: '))

if num1 % num2 == 0:
    print(f'{num1} é múltiplo de {num2}')
else:
    print(f'{num1} não é múltiplo de {num2}')
```

`a` é múltiplo de `b` quando a divisão de `a` por `b` não deixa resto: `a % b == 0`.
Cuidado com a ordem: 15 é múltiplo de 5 (`15 % 5 == 0`), mas 5 não é múltiplo de 15.

### Exercício 6

Uma empresa de vendas oferece aos clientes um desconto calculado em função do valor da
compra: **15%** se o valor for igual ou superior a R\$ 6999,00 e **5%** se for inferior.
Peça ao usuário o valor da compra e informe o desconto obtido e o valor final da compra.

```python
valor = float(input('Valor da compra: R$ '))

if valor >= 6999:
    percentual = 15
else:
    percentual = 5

desconto = valor * percentual / 100
print(f'Desconto de {percentual}%: R${desconto:.2f}')
print(f'Valor final: R${valor - desconto:.2f}')
```

### Exercício 8

Um caixa eletrônico tem somente cédulas de R\$ 50,00, R\$ 10,00 e R\$ 5,00. Obtenha o valor
do saque desejado pelo usuário e verifique se ele pode ser entregue. Se for possível,
mostre a quantidade **mínima** de cada tipo de cédula. Caso contrário, mostre a mensagem:
`Valor indisponível, dirija-se a outro caixa.`

```python
valor = int(input('Valor do saque: R$ '))

if valor % 5 != 0:
    print('Valor indisponível, dirija-se a outro caixa.')
else:
    cedulas_50 = valor // 50
    resto = valor % 50
    cedulas_10 = resto // 10
    resto = resto % 10
    cedulas_5 = resto // 5

    print(f'Cédulas de R$50: {cedulas_50}')
    print(f'Cédulas de R$10: {cedulas_10}')
    print(f'Cédulas de R$5: {cedulas_5}')
```

Para usar o mínimo de cédulas, começamos pela maior: tiramos o máximo possível de notas de
50, depois de 10 com o que sobrou e, por fim, de 5. Só é possível sacar valores múltiplos
de 5.

### Exercício 9

Um triângulo só existe se, e somente se, **cada um** dos seus lados for maior que o valor
absoluto (módulo) da diferença dos outros dois lados e menor que a soma dos outros dois.
Peça ao usuário as medidas dos lados de um triângulo e informe se existe ou não um
triângulo. Faça a verificação apenas se todos os valores forem positivos; caso algum seja
negativo, avise o usuário.

```python
a = float(input('Lado 1: '))
b = float(input('Lado 2: '))
c = float(input('Lado 3: '))

if a < 0 or b < 0 or c < 0:
    print('Os lados não podem ser negativos.')
else:
    lado_a_ok = abs(b - c) < a and a < b + c
    lado_b_ok = abs(a - c) < b and b < a + c
    lado_c_ok = abs(a - b) < c and c < a + b

    if lado_a_ok and lado_b_ok and lado_c_ok:
        print('Existe um triângulo com essas medidas.')
    else:
        print('Não existe um triângulo com essas medidas.')
```

A função `abs` devolve o valor absoluto de um número: `abs(-3)` é `3`. Guardar cada
condição em uma variável com nome claro (`lado_a_ok`) deixa o `if` mais fácil de ler.

### Exercício 10

Para ingressar em um curso de aviação, o candidato precisa atender **todas** as
condições abaixo. Obtenha a altura, a idade, as horas de voo e o peso de um candidato e
informe se ele está apto ou não para o curso.

- altura maior ou igual a 1,75 m;
- idade entre 22 e 40 anos (inclusive 22 e 40);
- horas de voo superior a 1600 horas;
- peso entre 65 e 95 kg (inclusive 65 e 95).

```python
altura = float(input('Altura (m): '))
idade = int(input('Idade: '))
horas_voo = float(input('Horas de voo: '))
peso = float(input('Peso (kg): '))

altura_ok = altura >= 1.75
idade_ok = idade >= 22 and idade <= 40
horas_ok = horas_voo > 1600
peso_ok = peso >= 65 and peso <= 95

if altura_ok and idade_ok and horas_ok and peso_ok:
    print('Candidato apto para o curso.')
else:
    print('Candidato não apto para o curso.')
```

Um erro comum é escrever `idade >= 22 or idade <= 40`. Com `or`, **qualquer** idade passa
no teste (por exemplo, 90 é maior que 22). Para "estar entre", as duas condições precisam
ser verdadeiras ao mesmo tempo: `and`.
