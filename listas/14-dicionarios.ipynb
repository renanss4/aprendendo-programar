## O que é um dicionário

Em uma lista, cada valor é acessado por uma posição numérica (o índice). Em um
**dicionário**, cada valor é acessado por uma **chave**, que normalmente é um texto ou um
número. É como um dicionário de verdade: você procura a palavra (chave) e encontra o
significado (valor).

````python
aluno = {
    'nome': 'Ana',
    'idade': 19,
    'curso': 'Sistemas de Informação',
}

print(aluno['nome'])
print(aluno['idade'])
````

## Adicionando, alterando e removendo

Atribuir a uma chave que não existe **cria** a entrada; atribuir a uma chave que já
existe **substitui** o valor. Para remover, use `del`:

````python
aluno = {'nome': 'Ana', 'idade': 19}

aluno['cidade'] = 'Florianópolis'   # cria
aluno['idade'] = 20                 # altera
del aluno['nome']                   # remove

print(aluno)
````

> As chaves de um dicionário são **únicas**: não existem duas entradas com a mesma chave.

## Consultando com segurança

Acessar uma chave que não existe causa `KeyError`. Para evitar o erro, teste antes com
`in` ou use o método `get`, que devolve um valor padrão quando a chave não existe:

````python
precos = {'café': 5.0, 'pão de queijo': 4.5}

if 'suco' in precos:
    print(precos['suco'])
else:
    print('Não temos suco.')

print(precos.get('café', 'indisponível'))
print(precos.get('suco', 'indisponível'))
````

## Percorrendo um dicionário

- `for chave in dicionario` (ou `dicionario.keys()`) percorre as chaves;
- `dicionario.values()` percorre os valores;
- `dicionario.items()` percorre os pares `(chave, valor)`.

````python
precos = {'café': 5.0, 'pão de queijo': 4.5, 'suco': 7.0}

for produto, preco in precos.items():
    print(f'{produto}: R$ {preco:.2f}')

total = 0
for preco in precos.values():
    total += preco
print('Total do cardápio:', total)
````

## Combinando dicionários e listas

Os valores de um dicionário podem ser de qualquer tipo, inclusive listas e outros
dicionários. Isso permite organizar dados mais complexos:

````python
turmas = {
    'manhã': ['Ana', 'Bruno'],
    'noite': ['Carla', 'Davi', 'Eva'],
}

turmas['noite'].append('Fábio')
for turno, alunos in turmas.items():
    print(f'{turno}: {len(alunos)} alunos')
````

## Erros comuns

- Acessar uma chave inexistente sem testar antes (`KeyError`).
- Esperar que o dicionário seja acessado por posição: `d[0]` procura a **chave** `0`.
- Usar uma lista como chave: as chaves precisam ser imutáveis (textos, números, tuplas).

### Exercício 1

Usando o dicionário `capitais`, crie uma função que recebe o nome de um estado e retorna
a sua capital. Caso o estado não esteja disponível, a função retorna o texto
`'indisponível'`. Faça duas versões da função:

- a) usando apenas o operador `in` e o acesso padrão a elementos de um dicionário;
- b) usando o método `get`.

````python
def capital_v1(estado):
    if estado in capitais:
        return capitais[estado]
    return 'indisponível'

def capital_v2(estado):
    return capitais.get(estado, 'indisponível')

print(capital_v1('Santa Catarina'), capital_v1('Texas'))
print(capital_v2('Amapá'), capital_v2('Texas'))
````

### Exercício 3

Usando os dicionários `regioes` e `siglas`, crie uma função que recebe o nome de uma
região e retorna uma lista com os estados que pertencem a ela. Caso a região não exista,
a função retorna `None`.

````python
def estados_da_regiao(regiao):
    if regiao not in regioes:
        return None

    estados = []
    for sigla in regioes[regiao]:
        estados.append(siglas[sigla])
    return estados

print(estados_da_regiao('Sul'))
print(estados_da_regiao('Atlântida'))
````

### Exercício 4

Usando os dicionários `regioes`, `siglas` e `capitais`, crie uma função que recebe o nome
de uma região e retorna uma lista com as capitais dos estados da região. Caso a região não
exista, a função retorna `None`.

````python
def capitais_da_regiao(regiao):
    if regiao not in regioes:
        return None

    lista_capitais = []
    for sigla in regioes[regiao]:
        estado = siglas[sigla]
        lista_capitais.append(capitais[estado])
    return lista_capitais

print(capitais_da_regiao('Sudeste'))
````

### Exercício 5

Mesma funcionalidade do exercício anterior, mas retornando um **dicionário** que tem o
estado como chave e, como valor, uma lista com a sigla e a capital.

````python
def dados_da_regiao(regiao):
    if regiao not in regioes:
        return None

    resultado = {}
    for sigla in regioes[regiao]:
        estado = siglas[sigla]
        resultado[estado] = [sigla, capitais[estado]]
    return resultado

print(dados_da_regiao('Sul'))
````

### Exercício 6

Usando todos os dicionários disponíveis, crie dicionário(s) para responder às perguntas
a seguir:

- a) Qual a população de cada região?
- b) Quais os estados e as populações de cada região?
- c) Qual a sigla, o estado, a região e a população de um estado?
- d) Quais os três estados com as menores populações? (Métodos úteis: `items` ou
  `values`; `sort` de lista para ordenar.)
- e) A mesma pergunta da letra d), mas usando a sua **própria função de ordenação**.

> Um dicionário organiza os dados por chaves. Por isso, dificilmente um único dicionário
> responde bem a todas as perguntas; mas, quanto melhor o projeto, menos dicionários serão
> necessários.

````python
# a) região -> população total
populacao_regiao = {}
for regiao, lista_siglas in regioes.items():
    total = 0
    for sigla in lista_siglas:
        total += populacoes[siglas[sigla]]
    populacao_regiao[regiao] = round(total, 2)
print('a)', populacao_regiao)

# b) região -> {estado: população}
estados_por_regiao = {}
for regiao, lista_siglas in regioes.items():
    estados_por_regiao[regiao] = {}
    for sigla in lista_siglas:
        estado = siglas[sigla]
        estados_por_regiao[regiao][estado] = populacoes[estado]
print('b)', estados_por_regiao['Sul'])

# c) estado -> todas as informações dele
ficha_estado = {}
for regiao, lista_siglas in regioes.items():
    for sigla in lista_siglas:
        estado = siglas[sigla]
        ficha_estado[estado] = {
            'sigla': sigla,
            'regiao': regiao,
            'capital': capitais[estado],
            'populacao': populacoes[estado],
        }
print('c)', ficha_estado['Santa Catarina'])

# d) três menores populações com sort: ordenamos pares [população, estado]
pares = []
for estado, populacao in populacoes.items():
    pares.append([populacao, estado])
pares.sort()
print('d)', pares[0][1], pares[1][1], pares[2][1])

# e) mesma coisa com uma ordenação feita à mão (ordenação por seleção)
def ordenar(lista):
    i = 0
    while i < len(lista):
        menor = i
        j = i + 1
        while j < len(lista):
            if lista[j] < lista[menor]:
                menor = j
            j += 1
        lista[i], lista[menor] = lista[menor], lista[i]
        i += 1

pares = []
for estado, populacao in populacoes.items():
    pares.append([populacao, estado])
ordenar(pares)
print('e)', pares[0][1], pares[1][1], pares[2][1])
````

Na letra d), cada par começa pela população, então `sort` ordena pela população. Na letra
e), a **ordenação por seleção** encontra o menor elemento e o coloca na primeira posição,
depois o segundo menor na segunda posição, e assim por diante. A linha
`lista[i], lista[menor] = lista[menor], lista[i]` troca dois elementos de lugar.

Na letra b), o enunciado original fala em "cidades", mas os dados disponíveis são de
estados; por isso usamos estados e suas populações.
