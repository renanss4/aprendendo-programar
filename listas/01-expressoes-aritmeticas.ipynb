## As três "divisões"

| Operador | Nome | Exemplo | Resultado |
| -------- | ---- | ------- | --------- |
| `/` | divisão | `15 / 6` | `2.5` |
| `//` | divisão inteira (quociente) | `15 // 6` | `2` |
| `%` | resto da divisão (módulo) | `15 % 6` | `3` |

O operador `%` aparece em muitos problemas: um número `n` é par quando `n % 2 == 0`, e
`n % 10` devolve o último dígito de `n`.

````python
print(15 / 6)
print(15 // 6)
print(15 % 6)
````

## Inteiros e reais

Python diferencia números **inteiros** (`int`, como `9`) de números **reais** (`float`,
como `9.0`). O resultado de `/` é sempre um `float`, mesmo quando a divisão é exata. Por
isso `9 / 1` mostra `9.0`.

> Em Python, o separador decimal é o **ponto**: escreva `1.75`, e não `1,75`.

## Precedência

Assim como na matemática, `**` é calculado primeiro, depois `*`, `/`, `//` e `%`, e por
último `+` e `-`. Use parênteses para mudar a ordem:

````python
print(10 + 5 * 30)     # 160: a multiplicação vem antes
print((10 + 5) * 30)   # 450: os parênteses mudam a ordem
print(9 ** (1 / 2))    # raiz quadrada de 9 = potência 1/2
````

## Erros comuns

- Usar vírgula como separador decimal: `1,75` não é o número 1,75 em Python.
- Esquecer os parênteses: `h * b + B / 2` é diferente de `h * (b + B) / 2`.
- Escrever `2a` em vez de `2 * a`: a multiplicação precisa ser explícita.

````python
print(15 / 6)
print(15 // 6)
print(15 % 6)
````

- `15 % 6` é o **resto** da divisão inteira: `3`.
